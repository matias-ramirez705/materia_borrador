## SLIDE 4: Estructura mínima y sintaxis de CREATE TABLE

```python
import psycopg2

# Conexión a base de datos
conn = psycopg2.connect(
    host='localhost',
    database='capacitaciones',
    user='postgres',
    password='password'
)
cur = conn.cursor()

# Estructura básica de una tabla
sql = """
CREATE TABLE nombre_tabla (
    nombre_columna tipo_dato [restricciones],
    ...
)
"""
print(sql)
```

## SLIDE 6: Ejemplo aplicado real - Sistema de gestión de RRHH

```python
# Tabla empleados con tipos de datos y restricciones esenciales
sql_empleados = """
CREATE TABLE IF NOT EXISTS empleados (
    id_empleado SERIAL PRIMARY KEY,
    rut VARCHAR(12) UNIQUE NOT NULL,
    nombre_completo VARCHAR(100) NOT NULL,
    fecha_ingreso DATE DEFAULT CURRENT_DATE,
    salario_base DECIMAL(10,2) CHECK (salario_base > 0)
)
"""

cur.execute(sql_empleados)
conn.commit()
print("Tabla empleados creada exitosamente")
```

## SLIDE 12: Tipos de datos numéricos

```python
# Ejemplo: Cálculo exacto de salarios en sistemas de nómina
sql_nomina = """
CREATE TABLE IF NOT EXISTS nomina (
    id_nomina SERIAL PRIMARY KEY,
    id_empleado INT NOT NULL,
    mes INT CHECK (mes BETWEEN 1 AND 12),
    salario_base DECIMAL(10,2) CHECK (salario_base >= 0),
    bonificacion DECIMAL(10,2) DEFAULT 0,
    descuentos DECIMAL(10,2) DEFAULT 0
)
"""

cur.execute(sql_nomina)
conn.commit()
print("Tabla nómina creada exitosamente")
```

## SLIDE 13: Tipos de datos de texto

```python
# Ejemplo: Almacenamiento de información de contacto
sql_contactos = """
CREATE TABLE IF NOT EXISTS contactos (
    id_contacto SERIAL PRIMARY KEY,
    rut CHAR(12) NOT NULL,
    nombre_completo VARCHAR(100) NOT NULL,
    correo_electronico VARCHAR(80) NOT NULL,
    direccion VARCHAR(255),
    ciudad VARCHAR(50)
)
"""

cur.execute(sql_contactos)
conn.commit()
print("Tabla contactos creada exitosamente")
```

## SLIDE 14: Tipos de datos de fecha y hora

```python
# Ejemplo: Control de asistencia y emisión de certificados
sql_asistencia_capacitacion = """
CREATE TABLE IF NOT EXISTS asistencia_capacitacion (
    id_asistencia SERIAL PRIMARY KEY,
    id_participante INT NOT NULL,
    id_sesion INT NOT NULL,
    fecha_asistencia DATE NOT NULL,
    hora_entrada TIME,
    hora_salida TIME,
    fecha_registro TIMESTAMP DEFAULT CURRENT_TIMESTAMP
)
"""

cur.execute(sql_asistencia_capacitacion)
conn.commit()
print("Tabla asistencia_capacitacion creada exitosamente")
```

## SLIDE 19: Ejemplos aplicados en contexto productivo - Caso 1

```python
# Control de registros de asistencia a capacitaciones
# Uso de CHECK para validar estados: Presente (P), Ausente (A), Justificado (J)
sql_asistencia = """
CREATE TABLE IF NOT EXISTS asistencia (
    id SERIAL PRIMARY KEY,
    id_participante INT NOT NULL,
    id_sesion INT NOT NULL,
    estado CHAR(1) DEFAULT 'P' CHECK (estado IN ('P','A','J'))
)
"""

cur.execute(sql_asistencia)
conn.commit()
print("Tabla asistencia creada exitosamente")
```

## SLIDE 19: Ejemplos aplicados en contexto productivo - Caso 2

```python
# Evitar duplicados en inscripciones
# Clave primaria compuesta para prevenir inscripciones duplicadas
sql_inscripciones = """
CREATE TABLE IF NOT EXISTS inscripciones (
    id_participante INT NOT NULL,
    id_curso INT NOT NULL,
    fecha_inscripcion DATE DEFAULT CURRENT_DATE,
    PRIMARY KEY (id_participante, id_curso)
)
"""

cur.execute(sql_inscripciones)
conn.commit()
print("Tabla inscripciones creada exitosamente")
```

## SLIDE 24: Actividad guiada - Modelado de tablas para gestión de capacitaciones
### Tabla 1: Participantes

```python
sql_participantes = """
CREATE TABLE IF NOT EXISTS participantes (
    id_participante SERIAL PRIMARY KEY,
    rut VARCHAR(12) UNIQUE NOT NULL,
    nombre_completo VARCHAR(100) NOT NULL,
    correo_electronico VARCHAR(80) NOT NULL,
    fecha_nacimiento DATE CHECK (fecha_nacimiento <= CURRENT_DATE)
)
"""

cur.execute(sql_participantes)
conn.commit()
print("Tabla participantes creada exitosamente")
```

## SLIDE 24: Actividad guiada - Tabla 2: Cursos

```python
sql_cursos = """
CREATE TABLE IF NOT EXISTS cursos (
    id_curso SERIAL PRIMARY KEY,
    nombre VARCHAR(100) NOT NULL,
    duracion_horas INT CHECK (duracion_horas > 0),
    modalidad VARCHAR(20) DEFAULT 'Online',
    fecha_inicio DATE
)
"""

cur.execute(sql_cursos)
conn.commit()
print("Tabla cursos creada exitosamente")
```

## SLIDE 24: Actividad guiada - Tabla 3: Inscripciones

```python
# Tabla que relaciona participantes y cursos
sql_inscripciones_capacitacion = """
CREATE TABLE IF NOT EXISTS inscripciones_capacitacion (
    id_participante INT NOT NULL,
    id_curso INT NOT NULL,
    fecha_inscripcion DATE DEFAULT CURRENT_DATE,
    PRIMARY KEY (id_participante, id_curso),
    FOREIGN KEY (id_participante) REFERENCES participantes(id_participante),
    FOREIGN KEY (id_curso) REFERENCES cursos(id_curso)
)
"""

cur.execute(sql_inscripciones_capacitacion)
conn.commit()
print("Tabla inscripciones_capacitacion creada exitosamente")
```

## Cerrar conexión

```python
cur.close()
conn.close()
print("Conexión cerrada")
```
