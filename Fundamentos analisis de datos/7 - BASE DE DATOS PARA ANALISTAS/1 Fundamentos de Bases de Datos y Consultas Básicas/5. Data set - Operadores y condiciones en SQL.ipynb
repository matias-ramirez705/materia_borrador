## 1. OPERADORES CONDICIONALES (Slides 4-11)

### Comparaciones Básicas (Slide 5)

```python
# Slide 5: Operadores condicionales básicos
print("SLIDE 5: Operadores Condicionales Básicos")
print("="*70)

operadores = pd.DataFrame({
    'Operador': ['=', '!=', '<>', '>', '<', '>=', '<='],
    'Significado': ['Igual a', 'Distinto de', 'Distinto de', 'Mayor que', 'Menor que', 'Mayor o igual', 'Menor o igual'],
    'Ejemplo': [
        "precio = 1000",
        "estado != 'Cancelado'",
        "estado <> 'Cancelado'",
        "edad > 18",
        "fecha < '2024-01-01'",
        "sueldo >= 500000",
        "puntos <= 80"
    ]
})

print(operadores.to_string(index=False))
```

### Comparaciones Igualdad (=) y Desigualdad (!=, <>)

```python
# Igualdad
df_igual = pd.read_sql_query(
    "SELECT nombre, servicio FROM clientes WHERE servicio = 'Internet'",
    conn
)
print("\n1. Igualdad (=): servicio = 'Internet'")
print(df_igual.to_string(index=False))

# Desigualdad
df_distinto = pd.read_sql_query(
    "SELECT nombre, servicio FROM clientes WHERE servicio <> 'Prueba' LIMIT 5",
    conn
)
print("\n2. Desigualdad (<>): servicio <> 'Prueba'")
print(df_distinto.to_string(index=False))
```

### Comparaciones de Magnitud (>, <, >=, <=)

```python
# Mayor que
df_mayor = pd.read_sql_query(
    "SELECT nombre, facturacion FROM clientes WHERE facturacion > 100000",
    conn
)
print("3. Mayor que (>): facturacion > 100000")
print(df_mayor.to_string(index=False))

# Menor o igual
df_menor_igual = pd.read_sql_query(
    "SELECT nombre, edad FROM clientes WHERE edad <= 30",
    conn
)
print("\n4. Menor o igual (<=): edad <= 30")
print(df_menor_igual.to_string(index=False))
```

### Comparaciones Alfanuméricas (Slide 7)

```python
# Slide 7: Comparar texto alfabéticamente
df_alfabetico = pd.read_sql_query(
    "SELECT nombre, ciudad FROM clientes WHERE nombre > 'L'",
    conn
)
print("SLIDE 7: Comparación Alfanumérica")
print("\nNombres posteriores a 'L' alfabéticamente:")
print(df_alfabetico.to_string(index=False))
```

### Comparaciones con Fechas (Slide 8)

```python
# Slide 8: Comparar fechas
df_fechas = pd.read_sql_query(
    "SELECT nombre, fecha_registro FROM clientes WHERE fecha_registro >= '2024-06-01'",
    conn
)
print("SLIDE 8: Comparación con Fechas")
print("\nClientes registrados desde 2024-06-01:")
print(df_fechas.to_string(index=False))
```

### Comparaciones con Expresiones (Slide 9)

```python
# Slide 9: Comparar con expresiones aritméticas
df_expresion = pd.read_sql_query(
    "SELECT nombre, facturacion, (facturacion * 0.1) as comision FROM clientes WHERE facturacion > 100000",
    conn
)
print("SLIDE 9: Comparación con Expresiones Aritméticas")
print("\nClientes con facturación > 100000 (con cálculo de comisión):")
print(df_expresion.to_string(index=False))
```

### Errores Comunes (Slide 12)

```python
errores_comparaciones = """
    SLIDE 12: ERRORES COMUNES EN COMPARACIONES
    ═════════════════════════════════════════════════════════════
    
    ✗ TRATAR NULL COMO VALOR COMÚN
      Problema: WHERE campo = NULL siempre devuelve falso
      Solución: Usar WHERE campo IS NULL
    
    ✗ OMITIR COMILLAS EN TEXTOS
      Incorrecto: WHERE ciudad = Santiago
      Correcto:   WHERE ciudad = 'Santiago'
    
    ✗ USAR FORMATO INCORRECTO EN FECHAS
      El formato debe coincidir con el motor SQL
    
    ✗ NO CONSIDERAR ZONAS HORARIAS
      En campos TIMESTAMP, ajustar correctamente el rango
    
    ✗ NO CONTROLAR NULL SIN INTENCIÓN
      Esto puede excluir registros válidos
      Usar: ... AND campo IS NOT NULL
"""

print(errores_comparaciones)
```

## 2. SENTENCIAS IN, NOT IN, BETWEEN (Slides 13-22)

### IN: Listas de Valores (Slide 14)

```python
# Slide 14: IN para listas
df_in = pd.read_sql_query(
    """SELECT nombre, ciudad, facturacion
    FROM clientes
    WHERE ciudad IN ('Santiago', 'Valparaíso', 'Concepción')""",
    conn
)

print("SLIDE 14: Operador IN para Listas")
print("\nClientes en ciudades: Santiago, Valparaíso, Concepción")
print(df_in.to_string(index=False))
print(f"\nTotal: {len(df_in)} registros")
```

### NOT IN: Exclusión de Valores (Slide 14)

```python
# Slide 14: NOT IN para exclusión
df_not_in = pd.read_sql_query(
    """SELECT nombre, servicio
    FROM clientes
    WHERE servicio NOT IN ('Prueba', 'Migración', 'Cancelación')""",
    conn
)

print("SLIDE 14: Operador NOT IN para Exclusión")
print("\nClientes sin servicios: Prueba, Migración, Cancelación")
print(df_not_in.to_string(index=False))
```

### BETWEEN: Rangos (Slide 18)

```python
# Slide 18: BETWEEN para rangos de edad
df_between_edad = pd.read_sql_query(
    """SELECT nombre, edad, servicio
    FROM clientes
    WHERE edad BETWEEN 30 AND 40""",
    conn
)

print("SLIDE 18: Operador BETWEEN para Rangos")
print("\n1. Rango de edad: 30 a 40 años")
print(df_between_edad.to_string(index=False))

# BETWEEN con facturación
df_between_fact = pd.read_sql_query(
    """SELECT nombre, facturacion
    FROM clientes
    WHERE facturacion BETWEEN 80000 AND 110000""",
    conn
)

print("\n2. Rango de facturación: $80,000 a $110,000")
print(df_between_fact.to_string(index=False))
```

### BETWEEN con Fechas (Slide 18-19)

```python
# BETWEEN con fechas
df_between_fecha = pd.read_sql_query(
    """SELECT nombre, fecha_registro
    FROM clientes
    WHERE fecha_registro BETWEEN '2024-06-01' AND '2024-07-15'""",
    conn
)

print("SLIDE 18-19: BETWEEN con Fechas")
print("\nRegistrados entre 2024-06-01 y 2024-07-15:")
print(df_between_fecha.to_string(index=False))
```

### Comparación: Operadores vs. BETWEEN (Slide 20)

```python
# Slide 20: Comparación equivalencia
comparacion_between = pd.DataFrame({
    'Sintaxis Tradicional': [
        'WHERE edad >= 30 AND edad <= 40',
        'WHERE facturacion >= 80000 AND facturacion <= 110000'
    ],
    'Sintaxis con BETWEEN': [
        'WHERE edad BETWEEN 30 AND 40',
        'WHERE facturacion BETWEEN 80000 AND 110000'
    ]
})

print("SLIDE 20: Equivalencia - Operadores vs. BETWEEN")
print("\n" + comparacion_between.to_string(index=False))
print("\n✓ BETWEEN es más legible y compacto")
```

## 3. OPERADORES LÓGICOS AND y OR (Slides 23-32)

### AND: Ambas Condiciones Deben Cumplirse (Slide 24)

```python
# Slide 24: AND - Ambas condiciones
df_and = pd.read_sql_query(
    """SELECT nombre, ciudad, edad, facturacion
    FROM clientes
    WHERE ciudad = 'Santiago' AND edad >= 30""",
    conn
)

print("SLIDE 24: Operador AND")
print("\nClientes en Santiago Y con edad >= 30:")
print(df_and.to_string(index=False))
```

### OR: Al Menos Una Condición Debe Cumplirse (Slide 27)

```python
# Slide 27: OR - Al menos una condición
df_or = pd.read_sql_query(
    """SELECT nombre, ciudad, edad
    FROM clientes
    WHERE ciudad = 'Valparaíso' OR ciudad = 'Concepción'""",
    conn
)

print("SLIDE 27: Operador OR")
print("\nClientes en Valparaíso O Concepción:")
print(df_or.to_string(index=False))
```

### Combinación de AND y OR (Slide 30)

```python
# Slide 30: Combinación con paréntesis
df_combinado = pd.read_sql_query(
    """SELECT nombre, ciudad, edad, facturacion
    FROM clientes
    WHERE (ciudad = 'Santiago' OR ciudad = 'Valparaíso')
      AND edad >= 30""",
    conn
)

print("SLIDE 30: Combinación de AND y OR")
print("\nClientes en (Santiago O Valparaíso) Y con edad >= 30:")
print(df_combinado.to_string(index=False))

print("\n⚠️ IMPORTANTE: Los paréntesis controlan el orden lógico")
print("   Sin paréntesis: A OR B AND C = A OR (B AND C)")
print("   Con paréntesis: (A OR B) AND C = resultado diferente")
```

## 4. CASO PRÁCTICO: Reportes Estratégicos (Slides 35-37)

Tres reportes solicitados por diferentes áreas de la empresa.

```python
print("SLIDE 35-37: Caso Práctico - Reportes Estratégicos")
print("="*70)

# Reporte Comercial
print("\n📊 REPORTE COMERCIAL (Slide 37a)")
print("Criterio: Clientes en Santiago, Antofagasta o Concepción")
print("          CON facturación > $100,000")

df_comercial = pd.read_sql_query(
    """SELECT nombre, ciudad, facturacion, servicio
    FROM clientes
    WHERE ciudad IN ('Santiago', 'Antofagasta', 'Concepción')
      AND facturacion > 100000""",
    conn
)
print(f"\nResultado: {len(df_comercial)} clientes")
print(df_comercial.to_string(index=False))
```

```python
# Reporte de Soporte
print("\n📊 REPORTE SOPORTE (Slide 37b)")
print("Criterio: Reclamos últimos 60 días")
print("          EXCLUYENDO servicios de 'Prueba'")

df_soporte = pd.read_sql_query(
    """SELECT nombre, fecha_reclamo, servicio
    FROM clientes
    WHERE fecha_reclamo IS NOT NULL
      AND fecha_reclamo >= date('2024-05-19')
      AND servicio NOT IN ('Prueba')""",
    conn
)
print(f"\nResultado: {len(df_soporte)} reclamos válidos")
print(df_soporte.to_string(index=False))
```

```python
# Reporte de Operaciones
print("\n📊 REPORTE OPERACIONES (Slide 37c)")
print("Criterio: Edad entre 30 y 45 años")
print("          EXCLUYENDO servicios 'Migración' o 'Cancelación'")

df_operaciones = pd.read_sql_query(
    """SELECT nombre, edad, ciudad, servicio
    FROM clientes
    WHERE edad BETWEEN 30 AND 45
      AND servicio NOT IN ('Migración', 'Cancelación')""",
    conn
)
print(f"\nResultado: {len(df_operaciones)} clientes operacionales")
print(df_operaciones.to_string(index=False))
```

## 7. ERRORES COMUNES RESUMIDOS

```python
errores_resumen = """
    ERRORES COMUNES Y CÓMO EVITARLOS
    ═════════════════════════════════════════════════════════════
    
    ❌ COMPARACIONES
    • NULL = NULL → ✅ NULL IS NULL
    • ciudad = Santiago → ✅ city = 'Santiago'
    • Formato fecha incorrecto → ✅ Validar formato YYYY-MM-DD
    
    ❌ IN / NOT IN
    • IN ('1', '2') en INTEGER → ✅ IN (1, 2)
    • Listas muy largas → ✅ Usar subconsultas
    • NULL en listas → ✅ Gestionar explícitamente
    
    ❌ BETWEEN
    • BETWEEN 40 AND 30 → ✅ BETWEEN 30 AND 40
    • Timestamp sin ajuste → ✅ Truncar hora/minuto
    • Campos con NULL no controlados → ✅ Usar OR campo IS NULL
    
    ❌ COMBINACIÓN AND/OR
    • A OR B AND C sin paréntesis → ✅ (A OR B) AND C
    • Confundir precedencia de operadores → ✅ Usar paréntesis siempre
    • Repetir columnas innecesariamente → ✅ Código limpio
"""

print(errores_resumen)
```
