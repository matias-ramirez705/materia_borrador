## Slide 5-9: Fundamentos Técnicos de la Conexión

```python
print("\n" + "="*80)
print("FUNDAMENTOS TÉCNICOS DE LA CONEXIÓN A DATOS EN POWER BI")
print("="*80)

print("""
CONEXIÓN A DATOS: MUCHO MÁS QUE UN PASO TÉCNICO

Conectarse a una fuente en Power BI es una DECISIÓN ARQUITECTÓNICA que define:
  • Cómo se estructura y carga la información
  • Qué transformaciones son posibles
  • Con qué frecuencia se actualiza
  • Cómo se asegura y controla
  • Cuál es el rendimiento final


MOTORES FUNDAMENTALES DE POWER BI
────────────────────────────────────────────────────────────────

1. MASHUP ENGINE
   Función: Evalúa y ejecuta consultas de Power Query
   Responsabilidades:
     • Conectar a fuentes de datos
     • Transformar y limpiar
     • Combinar múltiples tablas
     • Preparar datos para cargar
   Motor: M language (lenguaje de Power Query)

2. DATA MODEL ENGINE (VertiPaq)
   Función: Almacena y analiza datos en memoria
   Responsabilidades:
     • Compresión inteligente de datos
     • Cálculos rápidos con DAX
     • Relaciones entre tablas
     • Actualizaciones eficientes
   Beneficio: Análisis extremadamente rápido


CRITERIOS PARA ELEGIR UN MECANISMO DE CONEXIÓN
────────────────────────────────────────────────────────────────

1. ACTUALIZACIÓN REQUERIDA
   Pregunta clave: ¿Se necesita información en TIEMPO REAL?
   
   Si Tiempo Real → DirectQuery
   Si Periódica (diaria, semanal) → Importación
   
   Ejemplo:
     • ERP financiero: Tiempo real → DirectQuery
     • Reportes mensuales: Importación

2. TAMAÑO DEL CONJUNTO DE DATOS
   Pregunta clave: ¿Es manejable en MEMORIA?
   
   Si < 1GB → Importación (óptima)
   Si > 10GB → DirectQuery o Live Connection
   
   Limitación de Power BI Service:
     • Datasets importados: Típicamente < 1GB
     • DirectQuery: Ilimitado (consulta al origen)

3. CAPACIDAD DE TRANSFORMACIÓN
   Pregunta clave: ¿Requieren LIMPIEZA los datos?
   
   Si Sí → Importación (Power Query + DAX completo)
   Si No → DirectQuery (menos transformación)
   
   Ejemplo:
     • Excel desorganizado: Importación
     • Base de datos limpia y certificada: DirectQuery

4. CONTROL Y GOBERNANZA
   Pregunta clave: ¿La organización exige CENTRALIZACIÓN?
   
   Si Sí → Live Connection (modelo central)
   Si No → Importación (autonomía local)
   
   Implicancias:
     • Centralizado: Menos duplicidad, más control
     • Distribuido: Más autonomía, riesgo de inconsistencia

5. SEGURIDAD
   Pregunta clave: ¿Hay RESTRICCIONES legales o regulatorias?
   
   Si Sí → DirectQuery o Live Connection (no replica datos)
   Si No → Importación (copia local es segura)
   
   Casos sensibles:
     • Datos médicos: Restricción HIPAA
     • Datos financieros: Restricción PCI
     • Datos gubernamentales: Clasificación de seguridad

6. ARQUITECTURA DISPONIBLE
   Pregunta clave: ¿Se cuenta con INFRAESTRUCTURA?
   
   Recursos necesarios:
     • Gateways (para DirectQuery local)
     • Servidores (para Live Connection)
     • Modelos centralizados (para composición)


ARQUITECTURA DE CONEXIÓN EN POWER BI
────────────────────────────────────────────────────────────────

┌──────────────────────────────────┐
│ 1. CONEXIÓN                      │
│ Establece vínculo lógico con     │
│ origen (SQL, Excel, API, etc.)   │
└──────────────────────────────────┘
            ↓
┌──────────────────────────────────┐
│ 2. CREDENCIALES                  │
│ OAuth2, Puerta de enlace, Token  │
│ Autenticación segura             │
└──────────────────────────────────┘
            ↓
┌──────────────────────────────────┐
│ 3. DEFINICIÓN DE CONSULTAS       │
│ Power Query (Mashup)             │
│ Transformaciones, filtros        │
└──────────────────────────────────┘
            ↓
┌──────────────────────────────────┐
│ 4. EVALUACIÓN Y CARGA            │
│ En memoria (Importación) o       │
│ Delegada (DirectQuery)           │
└──────────────────────────────────┘
            ↓
┌──────────────────────────────────┐
│ 5. INTERACCIÓN                   │
│ Visualizaciones para usuario     │
│ Filtros, cálculos DAX            │
└──────────────────────────────────┘


ROL DE POWER BI GATEWAY EN CONEXIONES ON-PREMISES
────────────────────────────────────────────────────────────────

Cuando se utilizan fuentes locales (on-premises):
  • SQL Server en red interna
  • Archivos en servidor corporativo
  • Aplicaciones locales

Problem: Power BI Service está en la NUBE, no tiene acceso directo

Solución: Power BI GATEWAY
  • Intermediario entre Service (nube) y origen (local)
  • Requiere instalación en computadora/servidor local
  • Se conecta a Internet de forma segura
  • Permanentemente activo para actualizaciones automáticas
  • Encripción de credenciales


IMPACTO ORGANIZACIONAL DE CONEXIONES MAL DISEÑADAS
────────────────────────────────────────────────────────────────

Scenario 1: Dependencia de carpetas locales
  Problema:
    • Crear reporte que depende de archivo en C:\\Datos\\ventas.xlsx
    • Funciona en Desktop (local)
    • Se publica a Service (nube)
    • Service no puede acceder a C:\\ (no existe en la nube)
    • Actualización automática FALLA
  Impacto: Reportes desactualizados, confusión, desconfianza en datos

Scenario 2: Duplicidad de modelos
  Problema:
    • Equipo A crea modelo con datos de Ventas
    • Equipo B crea otro modelo con mismos datos de Ventas
    • Están en diferentes workspaces
    • Ambos importan desde SQL Server
    • Transformaciones DISTINTAS en cada uno
  Impacto: Números inconsistentes, conflictos organizacionales

Scenario 3: Mezcla sin documentación
  Problema:
    • Modelo con Importación + DirectQuery sin documentar
    • Relaciones bidireccionales entre ellos
    • Rendimiento impredecible
    • Difícil de mantener
  Impacto: Problemas de rendimiento, dificultad en troubleshooting


BUENAS PRÁCTICAS Y ERRORES COMUNES
────────────────────────────────────────────────────────────────

BUENAS PRÁCTICAS:
  ✓ Inventario documentado:
      - Qué fuentes usamos
      - Quién es responsable
      - Política de actualización
  ✓ Validar permisos ANTES de diseñar modelo
  ✓ Usar rutas COMPARTIDAS (no personales o temporales)
  ✓ Separar ambientes: Desarrollo ≠ Producción
  ✓ Evaluar factibilidad de actualización desde el inicio
  ✓ Documentar el mecanismo de conexión usado

ERRORES COMUNES:
  ✗ Conectar archivos locales y luego publicar sin adaptación
  ✗ Mezclar Import + DirectQuery sin control
  ✗ No considerar volumen de datos al modelar
  ✗ Depender de credenciales personales en conexiones compartidas
  ✗ Usar conexiones de red inestables como fuentes principales
  ✗ No documentar la lógica de conexión
  ✗ Confiar en acceso manual después de publicar
""")

print("="*80)
```

## Slide 12-15: Tipos de Mecanismos de Conexión

```python
print("\n" + "="*80)
print("TIPOS DE MECANISMOS DE CONEXIÓN EN POWER BI")
print("="*80)

print("""
1. IMPORTACIÓN (IMPORT)
────────────────────────────────────────────────────────────────

Funcionamiento:
  • Descarga UNA COPIA de los datos
  • Se almacenan en el modelo (en memoria - VertiPaq)
  • Requiere actualización periódica (programada)

VENTAJAS:
  ✓ Máximo rendimiento (VertiPaq es muy rápido)
  ✓ Compresión inteligente (1GB en BD = 100MB en Power BI)
  ✓ Permite transformaciones COMPLEJAS
  ✓ DAX COMPLETO disponible
  ✓ Relaciones avanzadas entre tablas
  ✓ Columnas calculadas y medidas
  ✓ Funciona OFFLINE después de cargar
  ✓ Compatible con casi TODAS las fuentes
  ✓ Reportes desconectados de fuente

LIMITACIONES:
  ✗ Datos pueden quedar DESACTUALIZADOS
  ✗ Requiere actualización automática bien configurada
  ✗ Alto consumo de recursos en carga inicial
  ✗ Tiempos largos de carga si volumen muy grande
  ✗ Requiere Gateway para fuentes on-premises
  ✗ Límite de tamaño (~1GB típico en Service)
  ✗ Mayor consumo de memoria en Power BI Service

CUÁNDO USAR:
  • Datos históricos o referencias
  • Ciclos de actualización definidos (ej: cierre diario)
  • Análisis complejos con DAX
  • Datos que necesitan limpieza
  • Combinación de múltiples fuentes

CASO PRODUCTIVO:
  Empresa retailer que importa diariamente:
    • Archivo Excel de ventas del día anterior (7 a.m.)
    • Base de datos de productos
    • Tabla de precios desde compartida
    → Combina todo, genera análisis de 24 horas
    → Disponible para equipo todo el día


2. DIRECTQUERY (CONSULTA DIRECTA)
────────────────────────────────────────────────────────────────

Funcionamiento:
  • NO importa datos
  • Cada interacción genera CONSULTA AL ORIGEN
  • Datos siempre ACTUALES
  • Sin almacenamiento local (casi)

VENTAJAS:
  ✓ TIEMPO REAL (refleja cambios inmediatamente)
  ✓ Ideal para datos en CONSTANTE CAMBIO
  ✓ Replicación de datos NO necesaria
  ✓ Cumple restricciones de seguridad (datos no se copian)
  ✓ Volumen de datos ILIMITADO
  ✓ Ideal para escaneos de conformidad (compliance)
  ✓ Menos consumo de memoria en Service

LIMITACIONES:
  ✗ Requiere conexión CONSTANTE a origen
  ✗ Rendimiento depende del servidor origen
  ✗ Consultas mal optimizadas = lentitud en UI
  ✗ Algunas transformaciones NO disponibles
  ✗ Funciones DAX LIMITADAS
  ✗ Sin columnas calculadas
  ✗ RLS (seguridad por fila) más compleja
  ✗ Si servidor origen cae → reporte no funciona
  ✗ Uso de gateway OBLIGATORIO para on-premises

CUÁNDO USAR:
  • Datos que cambian constantemente
  • Análisis financiero en tiempo real
  • Datos operacionales de ERP
  • Volúmenes muy grandes (> 10GB)
  • Restricciones de replicación de datos

CASO PRODUCTIVO:
  Banco que necesita reportes de SALDO ACTUAL:
    • Tabla de cuentas corrientes en SQL Server (10GB)
    • Actualizaciones cada minuto
    • DirectQuery → cada click genera SELECT
    → Usuario ve saldo EXACTO en el momento
    → Sin replicación de datos sensibles


3. LIVE CONNECTION (CONEXIÓN EN VIVO)
────────────────────────────────────────────────────────────────

Funcionamiento:
  • Similar a DirectQuery pero para MODELOS YA EXISTENTES
  • Conecta con Analysis Services o Dataset de Power BI
  • No crea nuevo modelo, reutiliza existente
  • Usa lógica de negocio centralizada

CÓMO FUNCIONA:
  Escenario:
    • Equipo de BI crea modelo centralizado en Analysis Services
    • Otros equipos NO crean modelos propios
    • Conectan con Live Connection a ese modelo
    • Todos usan mismas definiciones, medidas, seguridad

VENTAJAS:
  ✓ COHERENCIA TOTAL (todos ven lo mismo)
  ✓ Centralización de lógica de negocio
  ✓ Eliminación de duplicidad
  ✓ Actualizaciones automáticas en todos reportes
  ✓ Seguridad centralizada (RLS en un solo lugar)
  ✓ Ideal para governanza corporativa

LIMITACIONES:
  ✗ NO se pueden crear nuevas tablas
  ✗ NO se pueden crear nuevas medidas
  ✗ NO se puede editar Power Query
  ✗ Requiere infraestructura Analysis Services
  ✗ Dependencia total del modelo central
  ✗ Cambios en central afectan TODOS los reportes
  ✗ Requiere administradores especializados

CUÁNDO USAR:
  • Organizaciones grandes con múltiples equipos
  • Cuando existe modelo semántico central
  • Cuando gobernanza es prioridad
  • Cuando se requiere garantizar mismo número en toda empresa

CASO PRODUCTIVO:
  Gran empresa con 50 equipos de análisis:
    • Equipo central (3 personas) gestiona modelo en AS
    • 50 equipos se conectan con Live Connection
    • Todos crean reportes sobre MISMO modelo
    • Garantía: Número de "Ventas" es igual en todo reporte


4. MODELOS COMPUESTOS (COMPOSITE)
────────────────────────────────────────────────────────────────

Funcionamiento:
  • COMBINA múltiples modos en UN SOLO modelo
  • Ejemplo: Tabla importada + Tabla DirectQuery
  • Relaciones entre tablas de distintos modos

USO AVANZADO:
  Escenario:
    • Tabla de Productos → Importación (referencias estables)
    • Tabla de Ventas → DirectQuery (actualiza cada minuto)
    • Relación entre ambas → Automática
    → Reportes muestran productos actuales con ventas en tiempo real

VENTAJAS:
  ✓ FLEXIBILIDAD máxima
  ✓ Lo mejor de ambos mundos
  ✓ Optimización inteligente
  ✓ Escalabilidad en modelos complejos
  ✓ Priorización según criticidad

RIESGOS:
  ✗ COMPLEJIDAD muy alta
  ✗ Relaciones pueden comportarse inesperadamente
  ✗ Problemas de rendimiento si mal diseñado
  ✗ Difícil de mantener sin documentación
  ✗ Limitaciones en relaciones bidireccionales
  ✗ Requiere conocimiento AVANZADO

CUÁNDO USAR:
  • Modelos escalables de gran tamaño
  • Cuando exigen datos parcialmente en tiempo real
  • Cuando se requiere optimización extrema
  • Organizaciones con expertise técnico avanzado


COMPARATIVA RESUMIDA
────────────────────────────────────────────────────────────────

┌─────────────────┬──────────┬────────────┬──────────┬──────────┐
│ Criterio        │ Importa. │ DirectQry  │ Live     │ Compuest │
├─────────────────┼──────────┼────────────┼──────────┼──────────┤
│ Actualización   │ Programa │ Tiempo rel │ Tiempo   │ Mixto    │
│ Rendimiento     │ Óptimo   │ Depende    │ Bueno    │ Variable │
│ DAX Completo    │ Sí       │ No         │ Limitado │ Sí       │
│ Tamaño datos    │ < 1GB    │ Ilimitado  │ Ilimitado│ Ilimitado│
│ Transformación  │ Sí       │ Limitada   │ No       │ Sí       │
│ Complejidad     │ Baja     │ Media      │ Alta     │ Muy alta │
│ Gobernanza      │ Distrib. │ Distrib.   │ Central. │ Distrib. │
└─────────────────┴──────────┴────────────┴──────────┴──────────┘
""")

print("="*80)
```

## Slide 16: Buenas Prácticas Integradas

```python
print("\n" + "="*80)
print("BUENAS PRÁCTICAS INTEGRADAS")
print("="*80)

print("""
PRACTICA 1: DOCUMENTACIÓN
────────────────────────────────────────────────────────────────

Documentar el mecanismo por CADA TABLA:
  Ejemplo de inventario:
    Tabla                  Mecanismo       Frecuencia      Origen
    ─────────────────────────────────────────────────────────
    Productos              Importación     Diaria 8 a.m.   Excel SharePoint
    Ventas                 DirectQuery     Tiempo real     SQL Server
    Clientes               Importación     Semanal 10 p.m. BD CRM
    Cotización USD/CLP     Importación     Diaria 7 a.m.   API REST

  Beneficios:
    • Nuevo miembro entiende arquitectura rápido
    • Facilita troubleshooting
    • Auditoría y compliance
    • Identificar cuellos de botella


PRACTICA 2: SEPARACIÓN POR CAPAS
────────────────────────────────────────────────────────────────

Arquitectura de capas:

  CAPA 1: STAGING
    • Importación desde fuentes originales (tal como vienen)
    • Sin transformación
    • Uso: Auditoría, backup

  CAPA 2: MODELADO
    • Transformaciones complejas
    • Cálculos intermedios
    • Relaciones entre tablas
    • Uso: Análisis técnico

  CAPA 3: VISUALIZACIÓN
    • Reportes finales
    • Seguridad aplicada (RLS)
    • Presentación limpia
    • Uso: Consumo por usuarios finales

  Beneficio: Cambios en fuente no rompen reportes finales


PRACTICA 3: OPTIMIZACIÓN DE CONSULTAS
────────────────────────────────────────────────────────────────

Si usas DirectQuery:
  • EVITAR múltiples DirectQuery simultáneos
  • OPTIMIZAR consultas en el servidor origen
  • USAR índices apropiados
  • FILTRAR lo máximo en el servidor (no después)
  • LIMITAR las agregaciones en Power BI

Ejemplo MALO:
  SELECT * FROM VentasHistorica (1GB)
  → Filtrar en Power BI: Ventas > 1000
  → Problema: Traer 1GB a la nube para filtrar

Ejemplo BUENO:
  SELECT * FROM VentasHistorica
  WHERE Ventas > 1000
  → Problema resuelo en el servidor
  → Trae solo lo necesario


PRACTICA 4: CONTROL DE ACTUALIZACIONES
────────────────────────────────────────────────────────────────

Si usas Importación:
  • Programar actualización en horas baja demanda
  • No programar 10 modelos al mismo tiempo
  • Escalonar actualizaciones
  • Monitorear fallos de actualización
  • Alertas si actualización no completó

Ejemplo de escalonamiento:
  7:00 a.m. - Productos (5 min)
  7:10 a.m. - Clientes (10 min)
  7:25 a.m. - Ventas (20 min)
  7:50 a.m. - Consolidado (15 min)
  → No saturar servidor


PRACTICA 5: SEGURIDAD Y CREDENCIALES
────────────────────────────────────────────────────────────────

NONCA:
  ✗ Usar credenciales personales en conexiones compartidas
  ✗ Guardar contraseñas en plain text
  ✗ Confiar en credenciales de un empleado específico

SIEMPRE:
  ✓ Crear cuenta de servicio para conexiones
  ✓ Usar OAuth2 o Single Sign-On (SSO)
  ✓ Almacenar credenciales en Power BI (encriptado)
  ✓ Rotar credenciales periódicamente
  ✓ Documentar quién tiene acceso


PRACTICA 6: VALIDACIÓN Y TESTING
────────────────────────────────────────────────────────────────

Antes de publicar:
  1. Verificar que conexión funciona
  2. Probar en Desktop primero
  3. Validar números contra fuente original
  4. Simular actualización automática
  5. Probar en ambiente de staging (si existe)
  6. Obtener aprobación de dueño de datos


ERRORES FRECUENTES
────────────────────────────────────────────────────────────────

Error 1: Usar modelos compuestos sin estructura
  Problema:
    • Relaciones entre Import + DirectQuery sin reglas
    • Rendimiento impredecible
    • Difícil de debuggear
  Solución:
    • Documentar arquitectura
    • Limpiar pruebas en Development primero

Error 2: Ignorar límites de actualización
  Problema:
    • Power BI Service permite ~8 actualizaciones diarias
    • Crear 20 modelos que actualizan cada 30 min
    • Colas de actualización, caídas
  Solución:
    • Inventario de actualizaciones
    • Consolidar donde posible

Error 3: Asumir que DirectQuery = Import
  Problema:
    • Esperar mismo rendimiento
    • Crear medidas complejas que no funcionan
    • Ignorar limitaciones de funciones DAX
  Solución:
    • Documentación clara de limitaciones
    • Testing exhaustivo

Error 4: No considerar Gateway
  Problema:
    • Crear modelo que requiere Gateway
    • No instalar/configurar Gateway
    • Actualización automática falla silenciosamente
  Solución:
    • Evaluar Gateway desde el inicio
    • Documentar requisitos
    • Testear con Gateway activo
"""
    )

print("="*80)
```
