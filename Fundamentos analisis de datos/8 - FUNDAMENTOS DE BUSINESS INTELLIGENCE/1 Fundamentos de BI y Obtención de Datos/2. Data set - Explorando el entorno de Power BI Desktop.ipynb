## Slide 5-11: Vistas Principales de Power BI Desktop

```python
print("\n" + "="*80)
print("VISTAS PRINCIPALES DE POWER BI DESKTOP")
print("="*80)

print("""
1. VISTA DE REPORTES
─────────────────────────────────────────────────────────────

Propósito: Construcción visual del informe y comunicación de datos

Características clave:
  • Inserción de visualizaciones: gráficos, tablas, matrices, KPI, tarjetas
  • Personalización visual: títulos, colores, etiquetas, formato
  • Navegación interactiva: botones, bookmarks, drill-through, tooltips
  • Aplicación de filtros: a nivel visual, página o informe global
  • Múltiples páginas temáticas

Buenas prácticas:
  ✓ Definir jerarquía visual clara (título → resumen → detalle)
  ✓ Usar filtros y segmentadores para análisis ad hoc
  ✓ Aplicar temas corporativos para coherencia visual
  ✓ Nombrar visuales descriptivamente
  ✓ Usar tooltips explicativos en indicadores clave

Errores comunes:
  ✗ Incluir demasiadas visualizaciones en una sola página
  ✗ Crear interacciones que inducen a interpretaciones incorrectas
  ✗ No usar elementos interactivos coherentemente
  ✗ Colores y estilos inconsistentes


2. VISTA DE DATOS
─────────────────────────────────────────────────────────────

Propósito: Revisar, validar y explorar datos cargados en el modelo

Características clave:
  • Vista previa tabular completa de cada tabla del modelo
  • Inspección de columnas calculadas y campos derivados
  • Verificación del impacto de transformaciones (Power Query)
  • Detección de valores nulos, duplicados, rangos y distribuciones
  • Acceso a estadísticas básicas (min, max, promedio)

Buenas prácticas:
  ✓ Validar datos antes de crear visualizaciones
  ✓ Verificar correspondencia entre columnas clave (ID, códigos)
  ✓ Usar columnas calculadas temporales para validar reglas de negocio
  ✓ Confirmar tipos de datos y formatos

Errores comunes:
  ✗ Omitir validación y generar visualizaciones con datos incorrectos
  ✗ Asumir que Power Query corrigió todos los errores automáticamente
  ✗ No distinguir entre columnas calculadas y medidas
  ✗ Ignorar valores nulos que pueden afectar análisis


3. VISTA DE MODELO
─────────────────────────────────────────────────────────────

Propósito: Construir la arquitectura relacional del modelo de datos

Características clave:
  • Conexiones visuales entre tablas por relaciones
  • Visualización de cardinalidad (1:N, N:1, N:N, 1:1)
  • Tipo de relación y dirección de propagación de filtros
  • Activación/desactivación de relaciones
  • Identificación de relaciones activas vs inactivas
  • Organización espacial del lienzo

Buenas prácticas:
  ✓ Crear relaciones desde claves limpias y únicas
  ✓ Documentar las relaciones (notas, nombres claros)
  ✓ Agrupar visualmente tablas por temática
  ✓ Usar solo relaciones activas cuando sea posible
  ✓ Validar que la cardinalidad sea correcta

Errores comunes:
  ✗ Establecer relaciones incorrectas (ej. N:N sin control)
  ✗ Usar campos no únicos como claves de relación
  ✗ Omitir relaciones necesarias
  ✗ Crear relaciones circulares que confunden el filtrado
  ✗ No documentar el propósito de relaciones complejas
""")

print("="*80)
```

## Slide 21-22: Power Query - Motor de Transformación

```python
print("\n" + "="*80)
print("POWER QUERY: MOTOR DE TRANSFORMACIÓN DE DATOS")
print("="*80)

print("""
¿QUÉ ES POWER QUERY?

Power Query es el entorno dentro de Power BI donde se realiza la transformación
de datos antes de cargarlos al modelo. Combina interfaz gráfica con poder técnico.


¿QUÉ PERMITE HACER?

1. CONECTARSE A MÚLTIPLES FUENTES
   • Excel (.xlsx, .xls)
   • Archivos de texto (CSV, TSV, TXT)
   • Bases de datos (SQL Server, MySQL, PostgreSQL, Oracle)
   • Servicios web (APIs, JSON, XML)
   • Plataformas Microsoft (SharePoint, OneDrive, Azure)
   • Otros (PDF, OData, SAP, Salesforce, Google Analytics)

2. TRANSFORMAR DATOS
   Sin afectar las fuentes originales:
     • Renombrar columnas
     • Cambiar tipos de datos
     • Filtrar filas según criterios
     • Separar/dividir columnas
     • Pivotar o despivotar datos
     • Eliminar duplicados
     • Agrupar por categorías
     • Estandarizar formatos (fechas, monedas, textos)

3. AUTOMATIZAR EL FLUJO
   • Cada paso queda registrado en lista editable
   • Permite modificar o depurar el proceso
   • Reutilizar transformaciones en nuevos datos
   • Programar actualizaciones automáticas
   • Aplicar transformaciones a múltiples archivos

4. COMBINAR MÚLTIPLES TABLAS
   • Merge (unión horizontal): combinar por clave común
   • Append (unión vertical): apilar tablas con misma estructura
   • Crear relaciones durante la transformación

5. REALIZAR TRANSFORMACIONES AVANZADAS
   • Condicionales (si/entonces)
   • Funciones personalizadas (M language)
   • Parámetros para reutilización
   • Repetir pasos sobre múltiples archivos


FLUJO TÍPICO EN POWER QUERY

┌─────────────────────────────────────────┐
│ 1. CONECTAR A FUENTE                    │
│    └─ Excel, SQL, API, archivo local    │
├─────────────────────────────────────────┤
│ 2. VISUALIZAR DATOS                     │
│    └─ Revisar estructura inicial        │
├─────────────────────────────────────────┤
│ 3. APLICAR TRANSFORMACIONES             │
│    └─ Rename, Filter, Split, Merge      │
├─────────────────────────────────────────┤
│ 4. DOCUMENTAR PASOS                     │
│    └─ Cada paso es editable             │
├─────────────────────────────────────────┤
│ 5. CARGAR AL MODELO                     │
│    └─ Datos limpios listos para análisis│
└─────────────────────────────────────────┘


EJEMPLO APLICADO (CASO PRODUCTIVO):

Scenario: Cadena de supermercados importa diariamente archivos CSV de ventas
con nombres distintos por fecha (ej: ventas_2024_05_26.csv).

Sin Power Query:
  • Renombrar archivo manualmente cada día
  • Buscar y reemplazar headers inconsistentes
  • Eliminar filas de encabezado extra
  • Agrupar datos en hoja adicional
  • Proceso manual = ERROR + TIEMPO

Con Power Query:
  • Configurar conexión a carpeta con archivos
  • Crear transformación única que se aplica a TODOS
  • Parámetro para fecha automática
  • Merge de archivos automático
  • Programar actualización diaria
  • Resultado: Datos frescos, consistentes, sin intervención manual


BUENAS PRÁCTICAS EN POWER QUERY

✓ Crear una consulta por tabla (no todo en una)
✓ Documentar pasos complejos con comentarios
✓ Eliminar columnas innecesarias temprano
✓ Validar tipos de datos explícitamente
✓ Usar parámetros para reutilizar transformaciones
✓ Monitorear rendimiento de transformaciones
✓ Mantener copia de datos originales para auditoría

Errores comunes:
  ✗ Aplicar transformaciones innecesarias
  ✗ Afectar rendimiento con pasos redundantes
  ✗ No validar tipos de datos
  ✗ Crear lógica compleja en Power Query (mejor en Power Pivot)
  ✗ Olvidar actualizar pasos al cambiar fuentes
""")

print("="*80)
```

## Slide 23-26: Power Pivot - Motor de Modelado y Análisis

```python
print("\n" + "="*80)
print("POWER PIVOT: MOTOR DE MODELADO Y ANÁLISIS")
print("="*80)

print("""
¿QUÉ ES POWER PIVOT?

Power Pivot se activa una vez que los datos han sido transformados y cargados.
Permite construir el modelo analítico, aplicar reglas de negocio y generar
cálculos complejos con DAX (Data Analysis Expressions).


¿QUÉ PERMITE HACER?

1. CREAR MEDIDAS (KPIs y Cálculos Dinámicos)
   Medidas = Cálculos que se adaptan al contexto de filtros
   
   Ejemplos:
     • Total Ventas = SUM(Ventas[Monto])
     • Venta Promedio = AVERAGE(Ventas[Monto])
     • Cumplimiento = DIVIDE(SUM(Ventas), SUM(Meta))
     • Crecimiento YoY = ([Ventas Año Actual] - [Ventas Año Anterior]) / [Ventas Año Anterior]
     • Top 5 Productos = CALCULATE(SUM(Ventas), TOPN(5, Producto))

2. CREAR COLUMNAS CALCULADAS
   Columnas = Se evalúan fila por fila
   
   Ejemplos:
     • Clasificación Precio = IF(Precio > 100, "Premium", "Estándar")
     • Margen % = (Precio - Costo) / Precio
     • Mes Nombre = FORMAT(Fecha, "MMMM")

3. DEFINIR JERARQUÍAS
   Organizar datos en múltiples niveles:
     • Temporal: Año > Trimestre > Mes > Día
     • Geográfica: País > Región > Ciudad > Dirección
     • Producto: Categoría > Subcategoría > Producto
   
   Beneficio: Drill-down/up automático en visualizaciones

4. CONTROLAR COMPORTAMIENTO DEL MODELO
   • Definir relaciones entre tablas
   • Establecer dirección de filtros (una vía, ambas vías)
   • Marcar tablas como "No Visualizar" (tablas de configuración)
   • Ordenar columnas por otra columna (ej: mes por número)

5. CONSTRUIR ESTRUCTURAS EFICIENTES EN MEMORIA
   • Modelo tabular (VertiPaq): compresión inteligente
   • Optimización automática de almacenamiento
   • Caché en memoria para consultas rápidas
   • Escalabilidad: desde MB hasta GB sin degradación


DIFERENCIA: MEDIDAS vs COLUMNAS CALCULADAS

┌──────────────────────┬────────────────┬────────────────────────┐
│ Aspecto              │ MEDIDAS        │ COLUMNAS CALCULADAS    │
├──────────────────────┼────────────────┼────────────────────────┤
│ Evaluación           │ Contextual     │ Fila por fila          │
│ Tamaño de archivo    │ No afecta      │ Aumenta (se guardan)   │
│ Rendimiento          │ Rápido         │ Más lento con muchas   │
│ Uso de memoria       │ Bajo           │ Alto (data en disco)   │
│ Cuándo usar          │ Cálculos KPI   │ Clasificaciones        │
│ Ejemplo              │ Total Ventas   │ Categoría de precio    │
└──────────────────────┴────────────────┴────────────────────────┘


EJEMPLO DAX - MEDIDA COMPLEJA

# Tasa de Crecimiento Año a Año (YoY)

TasaCrecimiento = 
    DIVIDE(
        [Ventas Año Actual] - [Ventas Año Anterior],
        [Ventas Año Anterior]
    )

Explicación:
  • [Ventas Año Actual]: Medida que suma ventas del año actual
  • [Ventas Año Anterior]: Misma lógica pero año anterior
  • DIVIDE: Calcula la diferencia dividido por año anterior
  • Resultado: % de crecimiento


BUENAS PRÁCTICAS EN POWER PIVOT

✓ Mantener separadas las funciones:
    - Power Query: Limpieza y preparación
    - Power Pivot: Cálculos y lógica de negocio

✓ Documentar medidas:
    - Nombre claro: "Margen Neto %" no "MN"
    - Anotaciones explicativas
    - Fórmula disponible para revisar

✓ Evitar cálculos redundantes:
    - No calcular en Power Query si es mejor en Power Pivot
    - No repetir la misma lógica en medidas distintas

✓ Cargar solo datos necesarios:
    - Filtrar en Power Query
    - Evitar columnas irrelevantes
    - Mejorar rendimiento del modelo

✓ Verificar relaciones:
    - Antes de crear medidas complejas
    - Asegurar que filtros se propagan correctamente


ERRORES COMUNES

✗ Aplicar transformaciones innecesarias en Power Query
✗ Repetir cálculos en columnas Y medidas
✗ No validar relaciones entre tablas
✗ Nombrar genéricamente: "columna1", "medida2"
✗ Olvidar actualizar pasos al cambiar fuente de datos
✗ Crear jerarquías sin necesidad (complejidad innecesaria)
✗ Usar solo columnas calculadas para todo (ineficiente)
""")

print("="*80)
```

## Resumen Integrado

```python
print("\n" + "="*80)
print("RESUMEN - ENTORNO POWER BI DESKTOP")
print("="*80)

resumen = """
ARQUITECTURA DEL ENTORNO POWER BI DESKTOP

┌──────────────────────────────────────────────────────────┐
│ 1. INTERFAZ GRÁFICA                                      │
├──────────────────────────────────────────────────────────┤
│  Vistas:                                                 │
│    • Reportes: Construcción visual                       │
│    • Datos: Validación tabular                           │
│    • Modelo: Arquitectura relacional                     │
├──────────────────────────────────────────────────────────┤
│  Barras de herramientas:                                 │
│    • Inicio, Modelado, Insertar, Ver, Ayuda             │
├──────────────────────────────────────────────────────────┤
│  Paneles contextuales:                                   │
│    • Campos, Visualizaciones, Filtros, Propiedades      │
└──────────────────────────────────────────────────────────┘

┌──────────────────────────────────────────────────────────┐
│ 2. MOTORES INTERNOS                                      │
├──────────────────────────────────────────────────────────┤
│  Power Query:                                            │
│    • Transformación de datos                             │
│    • ETL (Extracción, Transformación, Carga)            │
│    • Automatización de procesos                          │
├──────────────────────────────────────────────────────────┤
│  Power Pivot:                                            │
│    • Modelado analítico                                  │
│    • Cálculos con DAX                                    │
│    • Jerarquías y relaciones                             │
└──────────────────────────────────────────────────────────┘

┌──────────────────────────────────────────────────────────┐
│ 3. FLUJO DE TRABAJO TÍPICO                               │
├──────────────────────────────────────────────────────────┤
│  1. Cargar datos → Power Query (Transformar)            │
│  2. Crear modelo → Power Pivot (Relacionar)             │
│  3. Diseñar visual → Vista de Reportes                   │
│  4. Validar → Vista de Datos                             │
│  5. Publicar → Power BI Service                          │
└──────────────────────────────────────────────────────────┘


PRINCIPALES CONCEPTOS

• VISTAS: Tres espacios de trabajo distintos con propósitos específicos
• HERRAMIENTAS: Barras de menú organizan funcionalidades por tema
• PANELES: Contextuales, se adaptan al contenido y selección
• POWER QUERY: Motor de ETL, transforma antes de modelar
• POWER PIVOT: Motor de análisis, crea estructura y cálculos
• DAX: Lenguaje para medidas complejas y dinámicas
• MEDIDAS: Cálculos contextuales, adaptativos a filtros
• COLUMNAS: Cálculos estáticos, fila por fila
• RELACIONES: Conexiones entre tablas que permiten filtrado cruzado
• JERARQUÍAS: Niveles que permiten drill-down/up en visualizaciones


CLAVES PARA DOMINAR POWER BI DESKTOP

1. Entender el propósito de cada vista
   → Reportes (visual), Datos (validación), Modelo (arquitectura)

2. Conocer dónde vive cada herramienta
   → Transformación: Power Query
   → Análisis: Power Pivot
   → Visualización: Vista de Reportes

3. Mantener separadas las responsabilidades
   → Limpieza en Power Query
   → Cálculos en Power Pivot
   → Formato en Reportes

4. Nombrar descriptivamente TODO
   → Facilita mantenimiento y colaboración
   → Evita confusión entre medidas/columnas

5. Documentar procesos complejos
   → Anotaciones en Power Query
   → Descripciones en medidas
   → Justificación de relaciones
"""

print(resumen)
print("="*80)
```
