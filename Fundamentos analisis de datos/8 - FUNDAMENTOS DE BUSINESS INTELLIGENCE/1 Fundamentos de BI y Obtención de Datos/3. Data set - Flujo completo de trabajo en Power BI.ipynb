```python
print("\n" + "="*80)
print("FASE 1: OBTENCIÓN DE DATOS")
print("="*80)

print("""
OBJETIVO: Identificar, acceder y conectar fuentes relevantes de datos.


FUENTES DE DATOS COMPATIBLES
────────────────────────────────────────────────────────────────

1. ARCHIVOS LOCALES
   • Excel (.xlsx, .xls)
   • CSV, TXT, JSON, XML
   • Ideal para: Datos pequeños, pruebas iniciales
   • Ventaja: Inmediato, sin requiere credenciales

2. BASES DE DATOS
   • SQL Server, PostgreSQL, Oracle, MySQL
   • Bases de datos en la nube: Azure SQL, Amazon RDS
   • Ideal para: Grandes volúmenes, datos operativos
   • Ventaja: Datos actuales, relaciones preexistentes

3. PLATAFORMAS EN LA NUBE
   • SharePoint Online, OneDrive, Google Drive
   • Azure Data Lake, Amazon S3
   • Ideal para: Colaboración, datos distribuidos
   • Ventaja: Acceso remoto, sincronización automática

4. SERVICIOS EN LÍNEA
   • APIs REST, Microsoft Exchange, Salesforce
   • Google Analytics, Facebook Ads, Twitter
   • Ideal para: Datos en tiempo real, análisis externo
   • Ventaja: Datos frescos, múltiples proveedores


EJEMPLO APLICADO (ESCENARIO REAL)
────────────────────────────────────────────────────────────────

Situación:
  Se requiere cruzar:
    • Archivo Excel de ventas regionales (local)
    • Información de tipo de cambio actual (API REST en línea)
    • Base de datos de productos (SQL Server corporativo)

Solución con Power BI:
  → Conectar Excel mediante archivo local
  → Conectar API mediante conector REST
  → Conectar SQL Server mediante credenciales
  → Consolidar todo en un único modelo


MODOS DE CONEXIÓN: IMPORTACIÓN vs DIRECTQUERY
────────────────────────────────────────────────────────────────

1. IMPORTACIÓN (Importación de datos)
   Funcionamiento:
     • Los datos se descargan y almacenan dentro del archivo .pbix
     • Operaciones sobre datos en memoria
     • Requiere actualización periódica

   Ventajas:
     ✓ Velocidad: Análisis rápido, sin latencia
     ✓ Funcionalidad completa: Todos los cálculos disponibles
     ✓ Offline: Funciona sin conexión a fuente

   Desventajas:
     ✗ Requiere actualizaciones programadas
     ✗ Tamaño limitado (capacidad de memoria)
     ✗ Datos no siempre actualizados en tiempo real

   Ideal para:
     • Datos históricos o que no cambian constantemente
     • Análisis de volúmenes pequeños-medianos
     • Equipos en la nube sin infraestructura local

2. DIRECTQUERY (Consulta directa)
   Funcionamiento:
     • Los datos se consultan en TIEMPO REAL desde la fuente
     • No se almacenan en el archivo .pbix
     • Cada consulta accede a la base de datos original

   Ventajas:
     ✓ Datos siempre actualizados
     ✓ Sin límite de tamaño del modelo
     ✓ Seguridad: Datos sensibles no se copian

   Desventajas:
     ✗ Requiere conexión constante
     ✗ Algunas funciones DAX no están disponibles
     ✗ Rendimiento dependiente de la fuente

   Ideal para:
     • Datos operacionales en tiempo real
     • Grandes volúmenes (bases de datos masivas)
     • Análisis que requieren precisión del dato actual

┌────────────────┬────────────────────┬──────────────────────┐
│ Criterio       │ IMPORTACIÓN        │ DIRECTQUERY          │
├────────────────┼────────────────────┼──────────────────────┤
│ Tamaño datos   │ < 1 GB (típico)    │ Ilimitado            │
│ Velocidad      │ Muy rápida         │ Depende de fuente    │
│ Actualización  │ Programada         │ Tiempo real          │
│ Offline        │ Sí                 │ No                   │
│ Funciones DAX  │ Todas              │ Limitadas            │
│ Costo          │ Bajo               │ Alto (requiere puerta)│
└────────────────┴────────────────────┴──────────────────────┘


VALIDACIÓN Y VERSIONAMIENTO
────────────────────────────────────────────────────────────────

✓ Asegurar que archivos/conexiones pertenezcan a versión ESTABLE
✓ Usar rutas RELATIVAS o URLs PÚBLICAS (no locales)
✓ Establecer reglas de validación:
    • Columnas esperadas presentes
    • Tipos de datos correctos
    • Rango de valores válidos
✓ Documentar el propósito y frecuencia de cada fuente


BUENAS PRÁCTICAS Y ERRORES COMUNES
────────────────────────────────────────────────────────────────

BUENAS PRÁCTICAS:
  ✓ Usar nombres claros en las conexiones
  ✓ Validar que fuente esté disponible y sea confiable
  ✓ Filtrar columnas/registros innecesarios desde origen
  ✓ Documentar credenciales y accesos
  ✓ Establecer calendario de actualización
  ✓ Probar conectividad periódicamente

ERRORES COMUNES:
  ✗ Conectar archivos locales que no funcionan en Service
  ✗ No comprobar credenciales o permisos de acceso
  ✗ Importar grandes volúmenes irrelevantes
  ✗ No validar la estructura de la fuente
  ✗ Usar conexiones no documentadas
  ✗ Ignorar cambios en la estructura de la fuente
""")

print("="*80)
```

## Slide 11-15: Preparación de Datos

```python
print("\n" + "="*80)
print("FASE 2: PREPARACIÓN DE DATOS")
print("="*80)

print("""
OBJETIVO: Limpiar, transformar y estructurar datos para análisis.
HERRAMIENTA: Power Query


OPERACIONES BÁSICAS EN POWER QUERY
────────────────────────────────────────────────────────────────

1. ELIMINACIÓN DE COLUMNAS INNECESARIAS
   • Reduce tamaño del modelo
   • Mejora rendimiento
   • Facilita legibilidad
   Ejemplo: Quitar códigos internos, IDs temporales

2. CAMBIO DE TIPO DE DATOS
   • Números: Integer, Decimal, Currency
   • Texto: String, Language
   • Fechas: Date, DateTime
   • Booleano: True/False
   Importancia: DAX y visualizaciones dependen de tipos correctos

3. DIVISIÓN DE COLUMNAS POR DELIMITADOR
   Ejemplo: "Santiago, Chile" → Ciudad: Santiago | País: Chile
   Beneficio: Permite análisis por componente

4. FILTRO DE VALORES NULOS O IRRELEVANTES
   • Elimina registros sin valor
   • Evita errores en cálculos
   • Mejora calidad del análisis

5. ELIMINACIÓN DE DUPLICADOS
   Ejemplo: Si hay ID_Producto repetido, mantener solo uno
   Métrica de éxito: Confirmar unicidad de claves


TRANSFORMACIONES AVANZADAS
────────────────────────────────────────────────────────────────

1. UNPIVOT (Despivotar)
   De: Tabla con columnas por producto (forma ancha)
   A: Tabla normalizada con filas por producto (forma larga)
   
   Estructura antes:          Estructura después:
   ┌─────┬───┬───┬───┐      ┌─────┬─────────┬───────┐
   │Mes  │P1 │P2 │P3 │      │Mes  │Producto │Valor  │
   ├─────┼───┼───┼───┤      ├─────┼─────────┼───────┤
   │Ene  │100│50 │75 │  →   │Ene  │P1       │100    │
   │Feb  │120│60 │80 │      │Ene  │P2       │50     │
   └─────┴───┴───┴───┘      │Ene  │P3       │75     │
                            └─────┴─────────┴───────┘

2. MERGE (Unión horizontal)
   Combina dos tablas por clave común
   Ejemplo: Tabla Ventas + Tabla Productos por ID_Producto

3. APPEND (Unión vertical)
   Apila filas de múltiples tablas con misma estructura
   Ejemplo: Ventas Ene + Ventas Feb + Ventas Mar

4. COLUMNAS CONDICIONALES
   Asigna categorías automáticamente
   Ejemplo: IF(Monto > 1000, "Alto", "Bajo")


CASO APLICADO (ESCENARIO PRODUCTIVO)
────────────────────────────────────────────────────────────────

Scenario: Empresa de retail recibe mensualmente archivos de ventas
con columnas diferentes para CADA PRODUCTO.

Archivo mes 1: Fecha | Tienda | Camisas | Pantalones | Zapatos
Archivo mes 2: Fecha | Tienda | Camisas | Pantalones | Zapatos

Problema manual:
  • Renombrar columnas mensualmente
  • Consolidar en una tabla de análisis
  • Aplicar tipos de datos
  • Tiempo: Horas de trabajo

Solución Power Query:
  1. Unpivot: Transformar columnas de productos a filas
  2. Append: Apilar todos los meses en una tabla
  3. Cambiar tipos: Estandarizar datos
  4. Automatizar: Aplicar a futuros meses
  → Resultado: Tiempo < 10 minutos por mes, sin errores manuales


DOCUMENTACIÓN Y TRAZABILIDAD
────────────────────────────────────────────────────────────────

Cada transformación queda registrada como un "Paso Aplicado".
Beneficios:
  • AUDITORÍA: Rastrear qué cambios se aplicaron
  • MODIFICACIÓN: Cambiar orden o lógica fácilmente
  • AUTOMATIZACIÓN: Reutilizar en futuras actualizaciones
  • DOCUMENTACIÓN: Transparencia para equipo

Buena práctica:
  Nombrar cada paso descriptivamente:
    ✓ "Eliminar columnas innecesarias"
    ✓ "Cambiar fecha a formato YYYY-MM-DD"
    ✗ "Paso 5"


BUENAS PRÁCTICAS Y ERRORES COMUNES
────────────────────────────────────────────────────────────────

BUENAS PRÁCTICAS:
  ✓ Nombrar claramente cada paso aplicado
  ✓ Aplicar filtros/limpieza desde origen cuando posible
  ✓ Comprobar tipos de datos tras transformaciones
  ✓ Validar que tabla final es coherente con objetivo
  ✓ Documentar transformaciones complejas
  ✓ Crear copia de datos originales para auditoría

ERRORES COMUNES:
  ✗ No revisar tipos de datos después de transformaciones
  ✗ Eliminar registros sin validar impacto
  ✗ Conservar columnas redundantes
  ✗ Aplicar transformaciones sin documentación
  ✗ Olvidar actualizar pasos al cambiar fuente
  ✗ Crear lógica de negocio en Power Query (mejor en DAX)
""")

print("="*80)
```

## Slide 17-21: Modelado de Datos y DAX

```python
print("\n" + "="*80)
print("FASE 3: MODELADO DE DATOS Y USO DE DAX")
print("="*80)

print("""
OBJETIVO: Estructurar datos lógicamente para análisis complejos.
HERRAMIENTA: Power Pivot + DAX


RELACIONES ENTRE TABLAS
────────────────────────────────────────────────────────────────

1. MODELO ESTRELLA (Star Schema)
   Centro: Tabla de HECHOS (transacciones)
   Alrededor: Tablas de DIMENSIONES (atributos)
   
   Ejemplo:
        ┌────────────────┐
        │   DIMENSIÓN    │
        │   Productos    │
        └────────────────┘
               △
               │
   ┌────────────┴────────────┐
   │     TABLA DE HECHOS     │
   │       Ventas            │
   └────────────┬────────────┘
               │
     ┌─────────┴─────────┐
     │                   │
  ┌──────────┐    ┌──────────┐
  │DIMENSIÓN │    │DIMENSIÓN │
  │Clientes  │    │  Fechas  │
  └──────────┘    └──────────┘

   Ventajas:
     ✓ Simple de entender
     ✓ Rendimiento optimizado
     ✓ Flexible para análisis

2. MODELO COPO DE NIEVE (Snowflake Schema)
   Similar a estrella pero dimensiones se dividen en sub-dimensiones
   Ventaja: Menor redundancia de datos
   Desventaja: Más complejo, relaciones adicionales


CLAVES Y RELACIONES
────────────────────────────────────────────────────────────────

CLAVE PRIMARIA (Primary Key)
  • Identifica únicamente cada fila en una tabla
  • Ejemplo: ID_Producto en tabla Productos
  • Requisito: No puede haber duplicados

CLAVE FORÁNEA (Foreign Key)
  • Vincula a clave primaria de otra tabla
  • Ejemplo: ID_Producto en tabla Ventas
  • Crea la relación entre tablas

RELACIONES COMUNES
  • 1:N (Uno a muchos): Cliente (1) tiene muchas Ventas (N)
  • N:1 (Muchos a uno): Múltiples Ventas pertenecen a un Cliente
  • N:N (Muchos a muchos): Uso raro, requiere tabla puente
  • 1:1 (Uno a uno): Raro, generalmente indica diseño mejorable


INTRODUCCIÓN A DAX: MEDIDAS Y COLUMNAS CALCULADAS
────────────────────────────────────────────────────────────────

1. MEDIDAS (Calculated Measures)
   Características:
     • Cálculos dinámicos
     • Se recalculan según filtros activos
     • Se usan en visualizaciones
     • No se almacenan en disco
   
   Ejemplo:
     Total Ventas = SUM(Ventas[Monto])
     Venta Promedio = AVERAGE(Ventas[Monto])
     Cumplimiento = DIVIDE(SUM(Ventas), SUM(Meta), 0)

2. COLUMNAS CALCULADAS (Calculated Columns)
   Características:
     • Se evalúan fila por fila
     • Se almacenan en el modelo
     • Aumentan tamaño del archivo
     • Útiles para clasificaciones
   
   Ejemplo:
     Margen = Ventas[PrecioVenta] - Ventas[Costo]
     Clasificación = IF(Monto > 1000, "Alto", "Bajo")


FUNCIONES DAX COMUNES
────────────────────────────────────────────────────────────────

FUNCIONES AGREGADAS:
  • SUM(columna): Suma de valores
  • AVERAGE(columna): Promedio
  • COUNT(columna): Cuenta de valores no vacíos
  • COUNTROWS(tabla): Cuenta de filas
  • MIN(columna): Valor mínimo
  • MAX(columna): Valor máximo

FUNCIONES DE FILTRADO:
  • CALCULATE(medida, filtro): Aplica filtro a medida
  • FILTER(tabla, condición): Filtra filas
  • ALL(tabla): Elimina todos los filtros

FUNCIONES DE TIEMPO:
  • TOTALYTD(medida, fecha): Total año hasta hoy
  • SAMEPERIODLASTYEAR(fechas): Mismo período año anterior
  • DATEADD(fecha, períodos, intervalos): Sumar períodos


CASO APLICADO: COMPARACIÓN AÑO A AÑO
────────────────────────────────────────────────────────────────

Requerimiento:
  Entregar reporte de ventas comparativas año a año con variación %

Medidas DAX necesarias:

  # Ventas del año actual
  Ventas Año Actual = SUM(Ventas[Monto])

  # Ventas del año anterior (usando tabla de fechas)
  Ventas Año Anterior = 
    CALCULATE(
      SUM(Ventas[Monto]),
      SAMEPERIODLASTYEAR(Fechas[Fecha])
    )

  # Variación porcentual
  Variacion % = 
    DIVIDE(
      [Ventas Año Actual] - [Ventas Año Anterior],
      [Ventas Año Anterior],
      0
    )

Resultado en visualización:
  → Usuario ve "2024: $150K vs 2023: $120K (+25%)"
  → Medidas se adaptan a filtros (por mes, región, etc.)


BUENAS PRÁCTICAS Y ERRORES COMUNES
────────────────────────────────────────────────────────────────

BUENAS PRÁCTICAS:
  ✓ Nombrar relaciones y medidas descriptivamente
  ✓ Usar medidas en lugar de columnas calculadas para métricas
  ✓ Validar que relaciones sean activas y necesarias
  ✓ Documentar lógica compleja de DAX
  ✓ Usar DIVIDE() para evitar divisiones por cero
  ✓ Probar medidas con filtros variados

ERRORES COMUNES:
  ✗ Crear relaciones circulares
  ✗ Usar campos no únicos como claves
  ✗ Realizar cálculos complejos sin filtrar contexto
  ✗ Relaciones bidireccionales innecesarias
  ✗ No documentar fórmulas DAX complejas
  ✗ Confundir medidas con columnas calculadas
""")

print("="*80)
```

## Slide 30-34: Reportería de Datos
