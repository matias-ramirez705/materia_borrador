## Slide 13: Flujo Operativo de un Sistema BI

```python
print("\nFLUJO OPERATIVO SISTEMA BI")
print("="*80)

flujo_bi = """
1. EXTRACCIÓN (Extract)
   └─ Se capturan datos desde múltiples fuentes:
      • Bases de datos relacionales (SQL Server, MySQL, Oracle)
      • APIs públicas y servicios web
      • Archivos (Excel, CSV, JSON, XML)
      • Sistemas transaccionales (ERP, CRM)
      • Sensores IoT

2. TRANSFORMACIÓN (Transform)
   └─ Se limpian, depuran y estandarizan los datos:
      • Eliminación de duplicados
      • Validación de tipos de datos
      • Estandarización de formatos (fechas, monedas)
      • Creación de columnas derivadas
      • Normalización de información

3. CARGA (Load)
   └─ Se almacenan en un repositorio analítico:
      • Data warehouse
      • Data lake
      • Modelos tabulares en memoria
      • Base de datos analítica

4. MODELADO DE DATOS
   └─ Se definen relaciones y estructura:
      • Relaciones entre tablas
      • Jerarquías (año > mes > día)
      • Medidas y KPIs
      • Columnas calculadas

5. ANÁLISIS Y VISUALIZACIÓN
   └─ Se crean reportes y dashboards:
      • Gráficos interactivos
      • Segmentadores y filtros
      • Indicadores clave (KPI)
      • Exportación de datos
"""

print(flujo_bi)
print("="*80)
```

## Slide 18-21: OLTP vs OLAP

```python
print("\n¿QUÉ ES OLTP?")
print("="*80)
print("""
OLTP = OnLine Transaction Processing

Sistemas optimizados para operaciones transaccionales del día a día.
Están diseñados para insertar, actualizar y eliminar datos de forma rápida.

Características:
  • Optimizados para velocidad en escritura (INSERT, UPDATE, DELETE)
  • Modelo de datos altamente normalizado
  • Transacciones pequeñas, frecuentes y concurrentes
  • Garantizan ACID (Atomicidad, Consistencia, Aislamiento, Durabilidad)
  • Usuarios: Personal operativo
  • Actualización: Tiempo real

Ejemplos:
  • Sistema POS (Punto de venta en supermercado)
  • Base de datos académica (Inscripciones)
  • Sistema de reservas (Vuelos, hoteles, clínicas)
  • ERP operacional
""")

print("\n¿QUÉ ES OLAP?")
print("="*80)
print("""
OLAP = OnLine Analytical Processing

Sistemas especializados en análisis multidimensional de grandes volúmenes.
Optimizados para lectura y análisis de datos históricos.

Características:
  • Optimizados para lectura intensiva y consultas complejas
  • Modelo de datos desnormalizado (modelo estrella o copo de nieve)
  • Permite agregaciones, filtros y exploración dinámica
  • Datos estructurados en cubos con dimensiones y medidas
  • Usuarios: Analistas, directivos
  • Actualización: Batch programado (diario, semanal)

Ejemplos de análisis:
  • Ventas mensuales por sucursal vs año anterior
  • Costos operativos por área y unidad de negocio
  • Comportamiento de clientes por campaña y ubicación
""")

print("\nCOMPARACIÓN OLTP vs OLAP")
print("="*80)

comparativa = pd.DataFrame({
    'Aspecto': [
        'Objetivo principal',
        'Tipo de consultas',
        'Modelo de datos',
        'Rendimiento',
        'Usuarios típicos',
        'Frecuencia de uso',
        'Actualización de datos',
        'Volumen de datos',
        'Complejidad'
    ],
    'OLTP': [
        'Operación diaria',
        'Simples, CRUD',
        'Normalizado',
        'Optimizado para escritura',
        'Personal operativo',
        'Constante',
        'Tiempo real',
        'Moderado',
        'Baja'
    ],
    'OLAP': [
        'Análisis estratégico',
        'Complejas, exploración y agrupación',
        'Desnormalizado',
        'Optimizado para lectura y agregación',
        'Analistas, directivos',
        'Eventual, intensiva',
        'Batch programado',
        'Muy grande',
        'Alta'
    ]
})

print(comparativa.to_string(index=False))
print("="*80)
```

## Slide 25-27: Fases del Trabajo BI (ETL, Modelado, Reporting)

```python
print("\nFASES DEL TRABAJO BI")
print("="*80)

print("""
FASE 1: ETL (Extract, Transform, Load)
────────────────────────────────────────

1. EXTRACCIÓN (Extract)
   • Se conectan múltiples fuentes de datos
   • Se define frecuencia de actualización
   • Se configura acceso seguro

2. TRANSFORMACIÓN (Transform)
   • Estandarización de formatos (fechas, monedas, unidades)
   • Eliminación de duplicados y valores erróneos
   • Creación de columnas derivadas (ej. clasificación por rangos)
   • Normalización o desnormalización según modelo posterior

3. CARGA (Load)
   • Inserción en repositorio analítico
   • Configuración de cargas periódicas
   • Validación de consistencia


FASE 2: MODELADO DE DATOS
────────────────────────────────────────

a) ESTRUCTURA DEL MODELO
   • Tablas de HECHOS: contienen registros numéricos (ventas, producción, costo)
   • Tablas de DIMENSIÓN: contienen atributos categóricos (cliente, fecha, producto)

b) RELACIONAMIENTO Y JERARQUÍAS
   • Relaciones 1:N entre tablas
   • Jerarquías de tiempo (año > trimestre > mes)
   • Jerarquías geográficas (país > región > ciudad)

c) CREACIÓN DE MEDIDAS Y COLUMNAS
   • Uso de expresiones DAX para construir indicadores
   • Ejemplo: Margen = SUM(Ventas[Ingreso]) - SUM(Ventas[Costos])

BUENAS PRÁCTICAS:
   ✓ Usar claves numéricas para relaciones (más eficientes)
   ✓ Nombrar medidas de forma comprensible
   ✓ Evitar relaciones circulares o ambigüedades


FASE 3: REPORTES Y VISUALIZACIONES (Reporting)
────────────────────────────────────────────────

a) TIPOS DE VISUALIZACIÓN
   • Comparaciones: gráficos de barra, líneas
   • Composición: gráficos de torta, anillos
   • Distribución: gráficos de dispersión
   • Tendencias: líneas temporales
   • Indicadores clave: tarjetas (cards), KPI visuales

b) ELEMENTOS INTERACTIVOS
   • Segmentadores (slicers) por dimensión
   • Filtros de página, visual y globales
   • Navegación entre páginas
   • Tooltips personalizados

c) PUBLICACIÓN Y CONSUMO
   • Publicación en Power BI Service
   • Definición de permisos (lectura, edición)
   • Programación de actualización automática
""")

print("="*80)
```

## Slide 39-40: ¿Qué es Power BI?

```python
print("\n¿QUÉ ES POWER BI?")
print("="*80)

print("""
Power BI es una herramienta de análisis de datos desarrollada por Microsoft que 
permite convertir datos en información visual, interactiva y accionable.

CARACTERÍSTICAS CLAVE:

1. Auto-servicio analítico
   └─ Permite a usuarios no técnicos explorar y analizar datos sin depender de TI

2. Visualizaciones interactivas
   └─ Reportes dinámicos y atractivos con múltiples formas de navegación

3. Modelado semántico
   └─ Crea relaciones entre tablas, define jerarquías y calcula métricas

4. Conectividad amplia
   └─ Accede a datos desde más de 100 fuentes (SQL, APIs, Azure, etc.)

5. Escalabilidad
   └─ Desde reportes personales hasta soluciones empresariales


VENTAJAS vs EXCEL AVANZADO:

┌─────────────────────┬──────────────────────┬──────────────────────┐
│   Característica    │   Excel Avanzado     │      Power BI         │
├─────────────────────┼──────────────────────┼──────────────────────┤
│ Capacidad de datos  │ Limitada a memoria   │ Optimizado en memoria│
│ Visualización       │ Manual, poco interact│ Interactiva, dinámica │
│ Automatización      │ Power Query, macros  │ Power Query + cargas  │
│ Compartir info      │ Archivos por correo  │ Publicación en nube   │
│ Escalabilidad       │ Limitada             │ Alta, corporativa     │
└─────────────────────┴──────────────────────┴──────────────────────┘
""")

print("="*80)
```

## Resumen de Conceptos Clave

```python
print("\nRESUMEN - CONCEPTOS CLAVE DE BI Y POWER BI")
print("="*80)

resumen = """
1. INTELIGENCIA DE NEGOCIO (BI)
   • Es el conjunto de herramientas, metodologías y procesos para convertir
     datos en decisiones.
   • Componentes: Fuentes, ETL, Repositorios, Procesamiento, Visualización

2. FLUJO BI
   • EXTRACCIÓN → TRANSFORMACIÓN → CARGA → MODELADO → VISUALIZACIÓN

3. OLTP vs OLAP
   • OLTP: Transacciones operativas, escritura optimizada, tiempo real
   • OLAP: Análisis estratégico, lectura optimizada, datos históricos

4. FASES DEL TRABAJO BI
   • ETL: Obtención y preparación de datos
   • MODELADO: Creación de estructura semántica
   • REPORTING: Visualización e interacción con datos

5. HERRAMIENTAS BI
   • Power BI: Integración Microsoft, comunidad activa, precio accesible
   • Tableau: Visualizaciones potentes, pero más costoso
   • Looker: Integración Google Cloud, enfoque centralizado
   • Qlik Sense: Análisis asociativo, interfaz compleja
   • Metabase: Código abierto, para proyectos pequeños

6. POWER BI - SUITE COMPLETA
   • Desktop: Desarrollo local, .pbix
   • Service: Publicación, colaboración, web
   • Mobile: Monitoreo en terreno, alertas

7. BENEFICIOS DE BI
   • Automatización de reportes
   • Decisiones basadas en datos
   • Monitoreo en tiempo real
   • Detección de anomalías
   • Capacidad predictiva

8. BUENAS PRÁCTICAS
   • Iniciar desde preguntas del negocio, no desde los datos
   • Separar OLTP de OLAP
   • Validar y documentar transformaciones
   • Usar modelo estrella para facilitar análisis
   • Diseñar reportes según público objetivo
"""

print(resumen)
print("="*80)
```
