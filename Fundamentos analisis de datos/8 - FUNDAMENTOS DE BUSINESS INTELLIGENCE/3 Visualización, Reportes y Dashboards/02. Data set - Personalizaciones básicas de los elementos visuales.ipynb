## SLIDE 14: Títulos Efectivos - Claridad y Contexto

### a) Función de los Títulos
Los títulos son la primera entrada cognitiva. Deben responder: **QUÉ, DÓNDE, CUÁNDO y POR QUÉ**

**❌ Mal:** "Gráfico de barras" | "Solicitudes por período"

**✓ Bien:** "Distribución mensual de solicitudes cerradas por comuna – Primer semestre 2024"

### b) Títulos Dinámicos con DAX

### Ejemplo 1: Título que se actualiza según filtro de mes

```python
# SLIDE 14 - Código DAX para Título Dinámico
# Uso en Power BI: Ir a Visualización > Título > Opciones de Título > fx (f(x) dinámico)

titulo_dinamico = '''
"Solicitudes cerradas en " & SELECTEDVALUE(Calendario[Nombre_Mes])
'''

print("TÍTULO DINÁMICO (DAX):")
print("="*60)
print(titulo_dinamico)
print("\nEfecto: El título cambia automáticamente según el mes seleccionado")
print("en el segmentador (ej: 'Solicitudes cerradas en Enero')")

# Ejemplo con variables
example_months = ["Enero", "Febrero", "Marzo", "Abril"]
for month in example_months:
    print(f"  → Si selecciona {month}: 'Solicitudes cerradas en {month}'")
```

### Ejemplo 2: Título con múltiples dimensiones

```python
# Título dinámico más complejo
titulo_avanzado = '''
"Solicitudes en " & SELECTEDVALUE(Comunas[Nombre_Comuna]) & 
" - " & SELECTEDVALUE(Calendario[Nombre_Mes]) & " " & 
FORMAT(SELECTEDVALUE(Calendario[Año]), "0000")
'''

print("TÍTULO DINÁMICO AVANZADO (DAX):")
print("="*60)
print(titulo_avanzado)
print("\nEjemplos de salida según filtros:")
examples = [
    ("La Florida", "Enero", 2024),
    ("Puente Alto", "Febrero", 2024),
    ("San Bernardo", "Marzo", 2024)
]
for comuna, mes, año in examples:
    print(f"  → 'Solicitudes en {comuna} - {mes} {año}'")
```

### c) Buenas Prácticas para Títulos
- Evitar abreviaturas técnicas si el público no es especializado
- Centrarlos visualmente o alinearlos con el diseño general
- Mantener coherencia con otros títulos del panel

## SLIDE 15: Leyendas - Claridad en la Categorización

### a) Rol de la Leyenda
Identifican el significado de colores o símbolos en un gráfico.

### b) Posicionamiento y Formato
- **Ubicación:** Parte superior o lateral derecha
- **Términos:** Consistentes con el modelo de datos
- **Eliminación:** Si existe solo una categoría → Eliminar la leyenda (evita redundancia)

**❌ Error común:** Gráfico con una única serie + leyenda redundante

**✓ Correcto:** Leyenda solo cuando hay múltiples series

```python
# SLIDE 15 - Ejemplos de Leyendas Correctas

# Caso 1: Gráfico de líneas con múltiples comunas (NECESITA leyenda)
leyenda_ejemplo_1 = {
    "tipo_grafico": "Línea",
    "series": ["La Florida", "Puente Alto", "San Bernardo", "Maipú"],
    "necesita_leyenda": True,
    "posicion": "Lateral derecha"
}

print("CASO 1: Gráfico con múltiples series")
print("="*60)
print(f"Tipo: {leyenda_ejemplo_1['tipo_grafico']}")
print(f"Series: {', '.join(leyenda_ejemplo_1['series'])}")
print(f"¿Necesita leyenda? {leyenda_ejemplo_1['necesita_leyenda']}")
print(f"Posición: {leyenda_ejemplo_1['posicion']}")

print("\n" + "="*60)

# Caso 2: Gráfico con una sola serie (NO necesita leyenda)
leyenda_ejemplo_2 = {
    "tipo_grafico": "Barras",
    "series": ["Total Solicitudes"],
    "necesita_leyenda": False,
    "razon": "Redundante: la serie es única y clara"
}

print("\nCASO 2: Gráfico con una sola serie")
print("="*60)
print(f"Tipo: {leyenda_ejemplo_2['tipo_grafico']}")
print(f"Series: {', '.join(leyenda_ejemplo_2['series'])}")
print(f"¿Necesita leyenda? {leyenda_ejemplo_2['necesita_leyenda']}")
print(f"Razón: {leyenda_ejemplo_2['razon']}")
```

## SLIDE 16-17: Etiquetas de Datos - Cuantificación Precisa

### a) Cuándo Usarlas
- Se comparan pocos valores
- Se desea enfatizar una diferencia mínima
- No existe un eje bien definido (tarjetas, embudos)

### b) Configuración Recomendada
- Ajustar formato numérico (porcentaje, moneda, miles)
- Aplicar formato condicional para destacar valores altos/bajos
- Activar solo para elementos clave (evitar saturación)

```python
# SLIDE 16-17 - Etiquetas de Datos Avanzadas

# Ejemplo: Etiqueta con formato personalizado
etiqueta_dax = '''
FORMAT(SUM(Solicitudes[DuracionHoras]), "#,##0.0") & " horas"
'''

print("ETIQUETA PERSONALIZADA (DAX) - Slide 17")
print("="*60)
print("Código DAX:")
print(etiqueta_dax)
print("\nEfecto visual en la etiqueta:")
ejemplos_etiquetas = [
    168.0,
    240.5,
    1500.75,
    5280.0
]
for valor in ejemplos_etiquetas:
    # Simular formato
    valor_formateado = f"{valor:,.1f}"
    print(f"  → {valor_formateado} horas")
```

```python
# Ejemplo: Formato condicional para etiquetas
# En Power BI: Seleccionar etiqueta > Formato condicional > Color

formateo_condicional = {
    "aplicacion": "Gráfico de columnas por comuna",
    "regla": "Destacar los 3 valores máximos",
    "metodo": "Formato condicional basado en percentiles"
}

print("FORMATO CONDICIONAL EN ETIQUETAS")
print("="*60)
for key, value in formateo_condicional.items():
    print(f"{key}: {value}")

print("\nEjemplo de aplicación:")
print("-" * 60)
print("Si las comunas tienen estos totales:")

comunas_datos = {
    "La Florida": 250,
    "Puente Alto": 180,
    "San Bernardo": 320,
    "Maipú": 195
}

# Ordenar descendente
comuna_ordenada = sorted(comunas_datos.items(), key=lambda x: x[1], reverse=True)

for i, (comuna, valor) in enumerate(comuna_ordenada, 1):
    # Top 3 resaltados
    if i <= 3:
        print(f"  {i}. {comuna}: {valor} ← RESALTADO (color rojo)")
    else:
        print(f"  {i}. {comuna}: {valor}")
```

## SLIDE 19-24: Uso de Colores, Bordes y Fondos

### a) Función del Color (Slide 20)
- **Enfatizar:** Destacar un KPI que supera la meta
- **Agrupar:** Mismo color para categoría repetida
- **Alertar:** Representar desviaciones o problemas
- **Separar:** Diferenciar series en gráfico comparativo

### b) Tipos de Paletas Cromáticas
- **Secuencial:** Un solo color (azul claro → azul oscuro) para magnitudes
- **Divergente:** Dos colores opuestos para extremos (verde = cumplimiento, rojo = incumplimiento)
- **Categórica:** Múltiples colores para categorías nominales

```python
# SLIDE 20 - Tipos de Paletas Cromáticas

import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
import numpy as np

fig, axes = plt.subplots(1, 3, figsize=(15, 3))
fig.suptitle('Tipos de Paletas Cromáticas en Visualización (Slide 20)', fontsize=14, fontweight='bold')

# 1. Paleta Secuencial
ax1 = axes[0]
ax1.set_title('Secuencial\n(Magnitudes: Claro → Oscuro)', fontweight='bold')
colores_secuencial = ['#E6F3FF', '#99CCFF', '#4C99FF', '#0052CC', '#003399']
for i, color in enumerate(colores_secuencial):
    rect = mpatches.Rectangle((i*0.2, 0), 0.2, 1, facecolor=color, edgecolor='black', linewidth=1)
    ax1.add_patch(rect)
ax1.set_xlim(0, 1)
ax1.set_ylim(0, 1)
ax1.set_xticks([])
ax1.set_yticks([])
ax1.text(0.5, -0.15, 'Bajo → Alto', ha='center', transform=ax1.transAxes)

# 2. Paleta Divergente
ax2 = axes[1]
ax2.set_title('Divergente\n(Extremos: Rojo ↔ Verde)', fontweight='bold')
colores_divergente = ['#D62828', '#F2A900', '#E6E6E6', '#A3C940', '#005288']
for i, color in enumerate(colores_divergente):
    rect = mpatches.Rectangle((i*0.2, 0), 0.2, 1, facecolor=color, edgecolor='black', linewidth=1)
    ax2.add_patch(rect)
ax2.set_xlim(0, 1)
ax2.set_ylim(0, 1)
ax2.set_xticks([])
ax2.set_yticks([])
ax2.text(0.5, -0.15, 'Neg ← Neutral → Pos', ha='center', transform=ax2.transAxes)

# 3. Paleta Categórica
ax3 = axes[2]
ax3.set_title('Categórica\n(Categorías distintas)', fontweight='bold')
colores_categorica = ['#005288', '#A3C940', '#F2A900', '#D62828', '#003049']
for i, color in enumerate(colores_categorica):
    rect = mpatches.Rectangle((i*0.2, 0), 0.2, 1, facecolor=color, edgecolor='black', linewidth=1)
    ax3.add_patch(rect)
ax3.set_xlim(0, 1)
ax3.set_ylim(0, 1)
ax3.set_xticks([])
ax3.set_yticks([])
ax3.text(0.5, -0.15, 'Comunas / Servicios', ha='center', transform=ax3.transAxes)

plt.tight_layout()
plt.show()

print("\nCódigos HEX de las paletas mostradas:")
print("Secuencial:", colores_secuencial)
print("Divergente:", colores_divergente)
print("Categórica:", colores_categorica)
```
