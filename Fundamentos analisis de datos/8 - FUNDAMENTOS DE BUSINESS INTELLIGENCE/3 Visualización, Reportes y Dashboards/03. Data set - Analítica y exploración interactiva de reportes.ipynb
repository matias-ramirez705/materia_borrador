### Medidas Analíticas Personalizadas con DAX (Slide 8)

```python
# SLIDE 8 - Medidas DAX Personalizadas

print("MEDIDAS ANALÍTICAS PERSONALIZADAS CON DAX (Slide 8)")
print("="*70)

# Ejemplo: Destacar valor máximo con color distinto
dax_ejemplo_1 = '''
ColorPersonalizado = 
IF(
    [CantidadSolicitudes] = CALCULATE(
        MAX([CantidadSolicitudes]),
        ALL(Comunas)
    ),
    "#FF5733",      // Rojo para máximo
    "#3366CC"       // Azul para otros
)
'''

print("\nEjemplo: Destacar valor máximo con color")
print("-" * 70)
print(dax_ejemplo_1)
print("\nEfecto: La comuna con mayor cantidad de solicitudes")
print("        se resalta en rojo, las demás en azul.")

# Ejemplo: Línea de meta
dax_ejemplo_2 = '''
DiferenciaMetaSolicitudes = 
[CantidadSolicitudes] - [MetaMensual]
'''

print("\n\nEjemplo: Calcular diferencia respecto a meta")
print("-" * 70)
print(dax_ejemplo_2)
print("\nEfecto: Valores positivos = superan meta")
print("        Valores negativos = no alcanzaron meta")
```

## SLIDE 17-21: Incorporación de Exploración en Profundidad

### ¿Qué es la Exploración en Profundidad? (Slide 17)

```python
# SLIDE 17 - Exploración en Profundidad

print("EXPLORACIÓN EN PROFUNDIDAD - CONCEPTOS (Slide 17)")
print("="*70)

exploracion_conceptos = {
    "Drill Down": {
        "descripcion": "Ir de un nivel general a un nivel más específico dentro de una jerarquía",
        "ejemplo": "De 'Año 2024' a 'Enero 2024' a 'Semana 1'"
    },
    "Drill Up": {
        "descripcion": "Volver del nivel específico al general",
        "ejemplo": "De 'Semana 1' a 'Enero 2024' a 'Año 2024'"
    },
    "Drill Through": {
        "descripcion": "Saltar a otra página del reporte con desglose detallado del dato seleccionado",
        "ejemplo": "Clic en 'Comuna: La Florida' → va a página 'Detalle Comuna'"
    },
    "Tooltips Personalizados": {
        "descripcion": "Mostrar información detallada al pasar el cursor sobre un dato",
        "ejemplo": "Pasar cursor sobre una barra → muestra promedio, meta, variación"
    }
}

for tipo, detalles in exploracion_conceptos.items():
    print(f"\n{tipo}:")
    print(f"  Descripción: {detalles['descripcion']}")
    print(f"  Ejemplo: {detalles['ejemplo']}")
```

### Exploración Jerárquica con Drill Down y Drill Up (Slide 18)

```python
# SLIDE 18 - Drill Down/Up

print("EXPLORACIÓN JERÁRQUICA CON DRILL DOWN Y DRILL UP (Slide 18)")
print("="*70)

print("""
Cómo configurarlo:
-" * 70)
pasos_drill = [
    "1. Asegúrate de que la jerarquía esté bien definida",
    "2. Agrega la jerarquía al eje del gráfico",
    "3. Activa las funciones de exploración en la parte superior",
    "4. Usa modo 'exploración por nivel' o 'exploración completa'"
]
for paso in pasos_drill:
    print(paso)

print("\n" + "="*70)
print("\nEjemplo Aplicado:")
print("-" * 70)
print("""
Jefatura de Operaciones - Análisis de Solicitudes:
  
  NIVEL 1 (Año):
  ├─ 2023: 5,250 solicitudes
  └─ 2024: 6,100 solicitudes  ← Usuario hace clic
  
  NIVEL 2 (Mes de 2024):
  ├─ Enero: 450 solicitudes
  ├─ Febrero: 500 solicitudes
  ├─ Marzo: 520 solicitudes    ← Usuario hace clic
  └─ ... (resto de meses)
  
  NIVEL 3 (Día de Marzo):
  ├─ Marzo 01: 18 solicitudes
  ├─ Marzo 02: 16 solicitudes
  ├─ Marzo 03: 22 solicitudes
  └─ ... (resto de días)
  
Permite análisis profundo sin abandonar la visualización.
""")
```

### Navegación Dirigida con Drill Through (Slide 19)

```python
# SLIDE 19 - Drill Through

print("NAVEGACIÓN DIRIGIDA CON DRILL THROUGH (Slide 19)")
print("="*70)

print("""
El Drill Through permite enviar al usuario a otra página del reporte,
enfocada en un detalle específico del dato seleccionado.

Cómo configurarlo:
-" * 70)
pasos_drill_through = [
    "1. Crea una nueva página del reporte (ej: 'Detalle Comuna')",
    "2. Agrega un visual con el campo que quieras analizar",
    "3. Arrastra al área 'Drill Through' el campo de contexto (ej: Comuna)",
    "4. Vuelve a página principal y verifica: clic derecho > 'Ir a detalle'"
]
for paso in pasos_drill_through:
    print(paso)

print("\n" + "="*70)
print("\nEjemplo de Flujo Drill Through:")
print("-" * 70)
print("""
Página Principal: 'Panel Solicitudes'
  ├─ Gráfico: Solicitudes por Comuna
  ├─ Usuario clic derecho en 'La Florida'
  └─ Selecciona 'Ir a detalle'
        ↓
Página Destino: 'Detalle Comuna'
  ├─ Título dinámico: "Solicitudes en: La Florida"
  ├─ Gráfico 1: Solicitudes por Tipo de Atención
  ├─ Gráfico 2: Evolución temporal
  ├─ Tabla: Detalle de solicitudes
  └─ Botón: 'Volver al Panel'
""")
```
