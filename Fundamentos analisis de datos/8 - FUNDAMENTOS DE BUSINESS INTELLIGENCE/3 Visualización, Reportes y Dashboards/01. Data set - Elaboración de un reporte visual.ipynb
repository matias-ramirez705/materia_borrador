## SLIDE 5-11: Creación de un Reporte - Conceptos Fundamentales

### ¿Qué es un Reporte Visual en Power BI? (Slide 5)
Un reporte visual es un **entorno interactivo** donde se integran:
- Gráficos
- Tablas
- Tarjetas
- Controles de segmentación

Para representar datos de forma visual, permitiendo:
- Comunicar hallazgos
- Hacer seguimiento de KPIs
- Brindar herramientas interactivas sin necesidad de escribir consultas

### Diferencia: Reporte vs Dashboard (Slide 6)

```python
# SLIDE 6 - Reporte vs Dashboard

diferencias = {
    "Reporte en Power BI": {
        "caracteristicas": [
            "Múltiples páginas",
            "Visualizaciones detalladas",
            "Profundiza en la información"
        ],
        "usuarios": "Analistas, equipos de operación",
        "ejemplo": "Páginas separadas: 'Solicitudes por Comuna' | 'Evolución Mensual' | 'Comparativo por Servicio'"
    },
    "Dashboard (Panel)": {
        "caracteristicas": [
            "Resumen de alto nivel",
            "Una sola vista",
            "Presentan indicadores clave"
        ],
        "usuarios": "Ejecutivos, tomadores de decisiones",
        "ejemplo": "KPIs principales sin análisis profundo"
    }
}

print("REPORTE VS DASHBOARD (Slide 6)")
print("="*70)
for tipo, detalles in diferencias.items():
    print(f"\n{tipo}:")
    print("-" * 70)
    print(f"Características: {', '.join(detalles['caracteristicas'])}")
    print(f"Usuarios: {detalles['usuarios']}")
    print(f"Ejemplo: {detalles['ejemplo']}")
```

### Proceso de Creación de un Reporte (Slide 7-8)

```python
# SLIDE 7-8 - Etapas del Proceso de Creación

proceso_creacion = {
    "1. Definir el objetivo": {
        "preguntas": [
            "¿Qué se quiere comunicar?",
            "¿A quién va dirigido?",
            "¿Qué decisiones debe facilitar?"
        ],
        "ejemplo": "Comunicar estado de solicitudes por comuna a directores municipales"
    },
    "2. Seleccionar indicadores clave (KPIs)": {
        "criterios": [
            "Alineados con objetivos estratégicos",
            "Disponibles en los datos",
            "Con frecuencia de actualización clara"
        ],
        "ejemplos_kpi": [
            "Tasa de resolución",
            "Cantidad de solicitudes por tipo",
            "Promedio de duración"
        ]
    },
    "3. Establecer estructura narrativa": {
        "orden_logico": "General → Comparativo → Detalle",
        "basarse_en": [
            "Geografía (comuna)",
            "Tiempo (mes, año)",
            "Servicio",
            "Estado"
        ],
        "ejemplo_estructura": "Página 1: Visión general | Página 2: Comparación intercomunal | Página 3: Evolución mensual"
    },
    "4. Insertar elementos visuales": {
        "asociaciones": {
            "Comparaciones": "Barras o líneas",
            "Proporciones": "Anillos o circulares",
            "Tendencias": "Líneas o áreas",
            "Magnitudes": "Tarjetas o medidores"
        }
    },
    "5. Aplicar filtros y segmentaciones": {
        "utilidad": "Permitir al usuario personalizar vistas",
        "ubicacion": "Parte superior o lateral",
        "ejemplos": ["Por comuna", "Por mes", "Por servicio"]
    },
    "6. Diseñar la experiencia de usuario (UX)": {
        "principios_gestalt": [
            "Proximidad",
            "Alineación",
            "Contraste",
            "Repetición"
        ],
        "recomendaciones": [
            "Evitar ruido visual",
            "Priorizar espacios en blanco",
            "Aplicar jerarquía tipográfica",
            "Usar colores consistentes con identidad institucional"
        ]
    },
    "7. Validar funcionalidad": {
        "pruebas": [
            "Pruebas de uso con personas clave",
            "Evaluar legibilidad",
            "Evaluar navegación",
            "Evaluar tiempo de carga",
            "Evaluar claridad de datos"
        ]
    }
}

print("PROCESO DE CREACIÓN DE UN REPORTE (Slides 7-8)")
print("="*70)
for etapa, detalles in proceso_creacion.items():
    print(f"\n{etapa}")
    print("-" * 70)
    for key, valor in detalles.items():
        if isinstance(valor, list):
            print(f"  {key}:")
            for item in valor:
                print(f"    • {item}")
        elif isinstance(valor, dict):
            print(f"  {key}:")
            for k, v in valor.items():
                print(f"    • {k}: {v}")
        else:
            print(f"  {key}: {valor}")
```
