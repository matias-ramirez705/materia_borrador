### Medidas DAX Sugeridas para KPIs

```python
# DAX para KPIs

print("\nMEDIDAS DAX PARA KPIs DEL DASHBOARD")
print("="*70)

dax_ejemplos = {
    "KPI: % Solicitudes Resueltas": """
PorcentajeResueltas = 
    DIVIDE(
        CALCULATE(COUNTROWS(Solicitudes), Solicitudes[Estado]="Cerrada"),
        COUNTROWS(Solicitudes),
        0
    ) * 100
    """,
    "KPI: Tiempo Promedio Resolución": """
TiempoPromedioResolucion = 
    AVERAGEX(Solicitudes, Solicitudes[DuracionHoras])
    """,
    "KPI: Total Solicitudes Activas": """
SolicitudesActivas = 
    CALCULATE(
        COUNTROWS(Solicitudes),
        Solicitudes[Estado]<>"Cerrada"
    )
    """,
    "KPI: Variación vs Mes Anterior": """
VariacionMesAnterior = 
    VAR MesActual = [CantidadSolicitudes]
    VAR MesAnterior = CALCULATE(
        [CantidadSolicitudes],
        DATEADD(Calendario[Fecha], -1, MONTH)
    )
    RETURN DIVIDE(MesActual - MesAnterior, MesAnterior, 0) * 100
    """
}

for kpi, formula in dax_ejemplos.items():
    print(f"\n{kpi}:")
    print(formula)
```
