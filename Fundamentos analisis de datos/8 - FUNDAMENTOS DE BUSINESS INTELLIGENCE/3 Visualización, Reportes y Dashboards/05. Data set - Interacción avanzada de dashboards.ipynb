## SLIDE 35-40: Definiendo Alertas en el Dashboard

### ¿Qué son las Alertas? (Slide 36)

```python
# SLIDE 36 - Definición Alertas

print("\n¿QUÉ SON LAS ALERTAS EN POWER BI? (Slide 36)")
print("="*70)

print("""
Alertas son notificaciones automáticas que:

✓ Se generan cuando una medida sobrepasa/queda debajo de umbral
✓ Permiten reacción proactiva ante desviaciones críticas
✓ Se configuran desde servicio Power BI online (no Desktop)

Requisitos:
  • Dashboard publicado en Power BI Service
  • Visualización: Card, KPI o Medidor
  • Permiso de edición y acceso al contenido
  • Datos actualizables (refresh habilitado)

Ventajas:
  ✓ Notificación automática por email
  ✓ Alertas visibles en Power BI
  ✓ No requiere revisión manual constante
  ✓ Acelera tiempo de respuesta

Casos de Uso:
  • % ejecución presupuestaria cae bajo 60%
  • Casos críticos sin resolución superan 100
  • Tiempo promedio atención excede 72 horas
  • Beneficiarios atendidos por debajo de meta mensual
""")
```

### Creación de una Alerta (Slide 37)

```python
# SLIDE 37 - Creación

print("\nCREACIÓN DE UNA ALERTA (Slide 37)")
print("="*70)

print("""
Pasos en Power BI Service (app.powerbi.com):

1. PUBLICAR EL REPORTE
   → Desde Power BI Desktop a Power BI Service

2. SELECCIONAR VISUALIZACIÓN
   → Ir a la Card/KPI/Medidor donde crear alerta

3. ACCEDER A ADMINISTRACIÓN DE ALERTAS
   → Esquina superior derecha: (...)
   → Seleccionar: "Administrar alertas"

4. AGREGAR ALERTA
   → Botón: "Agregar alerta"

5. CONFIGURAR PARÁMETROS:
   
   a) CONDICIÓN:
      ○ Mayor que (>)
      ○ Menor que (<)
      ○ Igual a (=)
   
   b) UMBRAL: Valor numérico crítico
      Ejemplo: 80% para cumplimiento de meta
   
   c) FRECUENCIA:
      ○ Notificación única
      ○ Recurrente (diaria, semanal, etc.)
   
   d) MÉTODO:
      ○ Email
      ○ Notificación en Power BI

6. GUARDAR
   → La alerta está activa y lista
""")
```

### Integración con Power Automate (Slide 38)

```python
# SLIDE 38 - Power Automate

print("\nALERTAS CON POWER AUTOMATE (Slide 38)")
print("="*70)

print("""
Power Automate permite flujos avanzados:

1. NOTIFICACIONES MÚLTIPLES
   → Enviar a múltiples personas/grupos
   → Automatizar distribución de alertas

2. REGISTRO EN SHAREPOINT
   → Documentar cada alerta activada
   → Mantener auditoría histórica

3. NOTIFICACIÓN EN TEAMS
   → Mensaje directo en canal Teams
   → Integración con comunicación organizacional

4. FLUJOS DE TRABAJO
   → Activar revisión o aprobación
   → Crear tickets de seguimiento
   → Notificar responsables automáticamente

Ejemplo de Flujo:

  PODER AUTOMATE TRIGGER: Alerta Power BI activada
    ↓
  ACCIÓN 1: Enviar email a gestor
    ↓
  ACCIÓN 2: Crear registro en SharePoint
    ↓
  ACCIÓN 3: Mensaje en Teams #operaciones
    ↓
  ACCIÓN 4: Crear ticket en sistema de tickets

Ventaja: Respuesta coordinada y documentada ante desviaciones.
""")
```
