# 9. Problema real: una columna numérica leída como texto

En el archivo existe al menos un valor de `cantidad` que no corresponde a un número escrito únicamente con dígitos.

Esto puede provocar que Pandas interprete toda la columna como `object`.

## Pregunta 5

Si una columna contiene:

```text

```

¿Cuál es el problema más probable?

A) Pandas convertirá automáticamente `"diez"` en `10`.  
B) La presencia de texto puede impedir que toda la columna sea interpretada como numérica.  
C) Pandas eliminará automáticamente la fila.  
D) La columna se convertirá en `bool`.


````python
# Identificamos los valores que no son numéricos
cantidad_numerica = pd.to_numeric(df["cantidad"], errors="coerce")

problemas = df[cantidad_numerica.isna()]

problemas
````

### Discusión

La instrucción:


Cuando encuentra un valor que no puede convertir, lo reemplaza por `NaN`.

> En una evaluación podrían preguntar qué ocurre con un valor no convertible cuando se utiliza `errors="coerce"`.

**Respuesta:** se convierte en `NaN`.


````python
df["cantidad"] = df["cantidad"].replace("diez", 10)
````

# 18. Desafío: lectura mejorada

Ahora queremos cargar el archivo considerando que:

- `fecha` debe ser interpretada como fecha.
- `cantidad` debería ser numérica.
- Si existe un valor no convertible, queremos detectarlo.

### Actividad

Completa el siguiente flujo:

1. Lee el archivo utilizando `parse_dates`.
2. Convierte `cantidad` a numérico.
3. Utiliza `errors="coerce"`.
4. Identifica las filas que generaron valores faltantes durante la conversión.


````python
archivo = "01. Apoyo ejercicio - Ventas mensuales.csv"

# Lectura de fecha
df = pd.read_csv(
    archivo,
    parse_dates=["fecha"]
)

# Conversión de cantidad
df["cantidad_numerica"] = pd.to_numeric(
    df["cantidad"],
    errors="coerce"
)

# Filas con problemas de conversión
errores_conversion = df[df["cantidad_numerica"].isna()]

display(errores_conversion)

# Revisar estructura
df.info()
````
